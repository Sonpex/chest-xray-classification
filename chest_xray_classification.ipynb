## I. Project Goal

The main goal of this project was to develop, optimize, and evaluate deep learning models for the automatic multi-class classification of chest X-ray images.

The implemented models classified chest X-rays into six categories:

* **COVID-19**
* **Emphysema**
* **Normal**
* **Pneumonia-Bacterial**
* **Pneumonia-Viral**
* **Tuberculosis**

The project aimed to achieve high classification performance while also considering model efficiency, including:

* the number of model parameters,
* the number of trainable parameters,
* model checkpoint size,
* compressed checkpoint size,
* inference latency,
* model sparsity,
* computational complexity.

The notebook compared custom convolutional neural networks trained from scratch with ImageNet-pretrained architectures. It also investigated the influence of data augmentation, MixUp, label smoothing, fine-tuning, model pruning, ensemble learning, and test-time augmentation.

## II. Problem Definition

Chest diseases may produce similar visual patterns in X-ray images, which makes reliable automatic classification challenging. This is particularly important for bacterial and viral pneumonia because their radiological characteristics can strongly overlap.

The project focused on building robust and generalizable convolutional neural network models that can distinguish between six lung disease classes.

The main research objectives:

* to compare custom CNN models trained from scratch with pretrained CNN architectures,
* to compare a plain scratch CNN with augmentation-assisted variants,
* to evaluate a more advanced optimized scratch CNN architecture,
* to evaluate the influence of training-time data augmentation,
* to examine the effect of MixUp regularization,
* to examine the influence of label smoothing,
* to evaluate test-time augmentation during inference,
* to compare several ImageNet-pretrained CNN backbones,
* to select the best transfer-learning model using validation Macro F1-score,
* to fine-tune only the best transfer-learning model,
* to investigate an additional extra-soft augmentation stage,
* to reduce model complexity through validation-controlled pruning,
* to improve classification robustness using validation-greedy ensemble learning,
* to compare predictive quality with computational efficiency,

Because the original X-ray images were grayscale, they were converted into three-channel images by replicating the grayscale channel. This makes them compatible with CNN architectures pretrained on ImageNet while avoiding the introduction of artificial colour information.

The images were normalized using the ImageNet mean and standard deviation.

## V. Applied Approaches

The following approaches were implemented and compared.

### Approach 1 – Simple Baseline CNN

Approach 1 used the same custom `SimpleCNN` architecture in two experimental variants.

#### Approach 1a – Baseline CNN Trained from Scratch

A custom convolutional neural network was trained from scratch using only deterministic image preprocessing.


It served as the clean reference model for the remaining experiments.

#### Approach 1b – Baseline CNN with Data Augmentation and TTA

The same `SimpleCNN` architecture was trained from scratch using a dedicated augmentation pipeline.

The training transformations included mild X-ray-safe operations such as:

* small rotations,
* small translations,
* limited scaling,
* slight brightness and contrast changes,
* mild Gaussian blur,
* random erasing.

During final evaluation, test-time augmentation was used to aggregate predictions from multiple versions of each test image.

This experiment evaluated whether augmentation-assisted training and TTA improve the generalization of the plain baseline architecture.

---

### Approach 2 – Optimized Baseline CNN with Augmentation, MixUp, Polish Stage, and TTA

A more advanced custom CNN architecture was trained from scratch.

Compared with the plain `SimpleCNN`, the optimized baseline included architectural and training improvements designed to increase classification quality.

The complete training procedure included:

* an optimized scratch CNN architecture,
* X-ray-safe data augmentation,
* weighted focal loss,
* label smoothing,
* MixUp regularization,
* a main training stage,
* a low-learning-rate polish stage,
* validation-based checkpoint selection,
* test-time augmentation during final evaluation.

---

### Approach 3 – Transfer Learning

Four ImageNet-pretrained CNN architectures were adapted to the six-class chest X-ray classification problem:

* ResNet50,
* DenseNet121,
* EfficientNet-B0,
* MobileNetV2.

Each transfer-learning experiment followed a staged training procedure that may included:

1. training a newly added classification head,
2. adapting the last pretrained backbone block,
3. applying a low-learning-rate polish stage.

The transfer-learning pipeline also used:

* mild X-ray-safe augmentation,
* a stronger classification head,
* focal loss,
* label smoothing,
* differential learning rates,
* validation-based early stopping.

---

### Approach 4 – Fine-Tuning the Best Transfer Model

Only the single best transfer-learning model selected in Approach 3 was fine-tuned.

The fine-tuning procedure included:

* loading the best transfer-learning checkpoint,
* unfreezing additional deeper backbone layers,
* using a smaller learning rate for the pretrained backbone,
* using a separate learning rate for the classification head,
* validation-based early stopping,
* validation-guarded checkpoint selection.

If fine-tuning did not improve validation Macro F1-score, the notebook kept the stronger original transfer-learning checkpoint.

---

### Approach 5 – Fine-Tuned Best Model with Extra-Soft Augmentation and TTA

The best fine-tuned model was further trained using a very gentle augmentation policy.

The extra-soft augmentation pipeline included:

* very small rotations,
* very small translations,
* minimal scaling,
* slight brightness and contrast variation,
* no aggressive random crop,
* reduced learning rates.

Only the last selected backbone block was adapted during this polish stage.

A validation guard compared the new checkpoint with the previous fine-tuned state. If the extra-soft augmentation stage reduces validation Macro F1-score, the previous checkpoint is preserved.

The final model was evaluated using test-time augmentation.

---

### Approach 6 – Quality-First Model Pruning

Global unstructured L1 pruning was applied to the strongest eligible single model.

Several mild pruning ratios were tested. For each pruning candidate:

* weights with the lowest absolute values were removed,
* pruning masks remained active during recovery training,
* a low-learning-rate recovery stage was performed,
* validation Macro F1-score was measured,
* final model sparsity was calculated.

The pruned candidate with the best validation Macro F1-score was selected. Sparsity was used only as a secondary criterion when candidates had very similar validation performance.

---

### Approach 7 – Validation-Greedy Ensemble with Test-Time Augmentation

The final ensemble was constructed using a validation-greedy selection procedure.

Candidate models were first ranked using validation Macro F1-score. Models were then added to the ensemble one at a time.

A candidate was retained only when its inclusion improves the ensemble validation Macro F1-score.

The selected models were combined using probability averaging.

During final test evaluation:

* several deterministic transformed versions of each image were generated,
* every selected ensemble model produced predictions for each version,
* all probabilities were averaged,
* the aggregated probabilities were used to obtain the final class prediction.

This approach combined:

* validation-only model selection,
* greedy ensemble construction,
* probability averaging,
* test-time augmentation.

---

## VI. Training and Model-Selection Protocol

To ensure a consistent and fair comparison, all approaches used the same general evaluation utilities and dataset splits.

The main experimental rules were:

* the training subset was used for parameter optimization,
* the validation subset was used for early stopping and model selection,
* Macro F1-score was the main model-selection metric,
* the test subset was used only for final evaluation,
* the best transfer model was selected before fine-tuning,
* only one transfer model was fine-tuned,
* fine-tuning and extra-soft augmentation used validation guards,
* pruning candidates were selected using validation performance,
* ensemble members were selected using validation predictions,
* test results were not used to decide which models should be selected.

# 2. Dataset Download and Structure Detection

The ChestX6 dataset was downloaded automatically from Kaggle using `kagglehub`. Because the downloaded archive structure could vary, the code recursively searched for the directory containing the predefined `train`, `val`, and `test` subsets.

After the correct dataset root had been detected, the paths to all three subsets were stored and displayed. This ensured that the remaining notebook sections used the correct dataset directories without requiring manual path configuration.


```python
raw_dataset_path = Path(kagglehub.dataset_download(CONFIG["dataset_slug"]))
print("Raw Kaggle dataset path:", raw_dataset_path)

def find_split_root(root: Path) -> Path:
    candidates = [root] + [p for p in root.rglob("*") if p.is_dir()]
    for p in candidates:
        if all((p / split).is_dir() for split in ["train", "val", "test"]):
            return p
    raise FileNotFoundError("Could not find a folder containing train, val and test subdirectories.")

DATA_ROOT = find_split_root(raw_dataset_path)
SPLITS = ["train", "val", "test"]
print("Detected dataset root:", DATA_ROOT)
for split in SPLITS:
    print(split, "->", DATA_ROOT / split)
```

# 3. Dataset and EDA

This section explored the dataset structure, class distribution, image properties, and data integrity. Exploratory data analysis was performed before model training to identify possible imbalance, inconsistent image characteristics, and duplicate leakage between dataset subsets.


All supported image files were collected from the training, validation, and test directories. A structured table was created with the image path, dataset split, and class label, and the number of images in each class and subset was calculated.

```python
IMG_EXTENSIONS = {".png", ".jpg", ".jpeg", ".bmp", ".tif", ".tiff", ".webp"}

records = []
for split in SPLITS:
    split_dir = DATA_ROOT / split
    for class_dir in sorted([p for p in split_dir.iterdir() if p.is_dir()]):
        for file_path in class_dir.rglob("*"):
            if file_path.suffix.lower() in IMG_EXTENSIONS:
                records.append({
                    "split": split,
                    "class": class_dir.name,
                    "path": str(file_path),
                })

files_df = pd.DataFrame(records)
class_names = sorted(files_df["class"].unique().tolist())
print("Number of images:", len(files_df))
print("Classes:", class_names)
display(files_df.head())

counts = (
    files_df.groupby(["split", "class"])
    .size()
    .rename("n")
    .reset_index()
    .sort_values(["split", "class"])
)
display(counts)

pivot_counts = counts.pivot(index="class", columns="split", values="n").fillna(0).astype(int)
display(pivot_counts)
```

The class distribution was visualized separately for the training, validation, and test subsets. The total number of images in each subset was also presented to assess dataset balance and verify the predefined split proportions.

MD5 hashes were calculated for all images to detect identical files within and across the dataset subsets. Cross-split duplicates were removed from lower-priority subsets while preserving the test set first, followed by the validation and training sets, and a final verification confirmed that no duplicate leakage remained.

```python
def canonical_path(path):
    return str(Path(path).resolve())

def md5_file(path, chunk_size=1024 * 1024):
    h = hashlib.md5()
    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(chunk_size), b""):
            h.update(chunk)
    return h.hexdigest()

duplicate_scan = files_df.copy()
duplicate_scan["canonical_path"] = duplicate_scan["path"].map(canonical_path)
duplicate_scan["md5"] = [md5_file(p) for p in tqdm(duplicate_scan["path"], desc="MD5 scan")]
duplicate_summary = (
    duplicate_scan.groupby("md5")
    .agg(
        n=("path", "count"),
        splits=("split", lambda x: sorted(set(x))),
        classes=("class", lambda x: sorted(set(x))),
    )
    .query("n > 1")
    .sort_values("n", ascending=False)
)

print(f"Duplicate groups found: {len(duplicate_summary)}")
display(duplicate_summary.head(10))

cross_split_duplicates = duplicate_summary[
    duplicate_summary["splits"].apply(lambda x: len(x) > 1)
]
print(f"Duplicate groups crossing train/val/test splits: {len(cross_split_duplicates)}")
display(cross_split_duplicates.head(10))

SPLIT_PRIORITY = {"train": 1, "val": 2, "test": 3}
EXCLUDED_IMAGE_PATHS = set()

if not cross_split_duplicates.empty:
    duplicate_records = duplicate_scan[duplicate_scan["md5"].isin(cross_split_duplicates.index)].copy()
    duplicate_records["split_priority"] = duplicate_records["split"].map(SPLIT_PRIORITY)

    for md5_hash, group in duplicate_records.groupby("md5"):
        keep_priority = group["split_priority"].max()
        lower_priority_paths = group.loc[
            group["split_priority"] < keep_priority,
            "canonical_path",
        ]
        EXCLUDED_IMAGE_PATHS.update(lower_priority_paths.tolist())

excluded_duplicates_df = duplicate_scan[
    duplicate_scan["canonical_path"].isin(EXCLUDED_IMAGE_PATHS)
].sort_values(["md5", "split", "class", "path"])

print(f"Images excluded to prevent cross-split duplicate leakage: {len(EXCLUDED_IMAGE_PATHS)}")
if not excluded_duplicates_df.empty:
    display(excluded_duplicates_df[["split", "class", "path", "md5"]].head(20))

clean_duplicate_scan = duplicate_scan[
    ~duplicate_scan["canonical_path"].isin(EXCLUDED_IMAGE_PATHS)
].copy()
remaining_cross_split_duplicates = (
    clean_duplicate_scan.groupby("md5")
    .agg(n=("path", "count"), splits=("split", lambda x: sorted(set(x))), classes=("class", lambda x: sorted(set(x))))
    .query("n > 1")
)
remaining_cross_split_duplicates = remaining_cross_split_duplicates[
    remaining_cross_split_duplicates["splits"].apply(lambda x: len(x) > 1)
]

if not remaining_cross_split_duplicates.empty:
    display(remaining_cross_split_duplicates.head(10))
    raise RuntimeError("Cross-split duplicate leakage remains after filtering.")

DATASET_LEAKAGE_STATUS = (
    "No cross-split duplicate leakage detected."
    if not EXCLUDED_IMAGE_PATHS
    else f"Excluded {len(EXCLUDED_IMAGE_PATHS)} lower-priority duplicate image(s) before modeling."
)
print(DATASET_LEAKAGE_STATUS)
```

# 5. Data loaders

This section created the training, validation, and test data loaders. Leakage-risk duplicates were excluded, class coverage was verified, and an optional weighted sampler was used to reduce the influence of class imbalance during training.

Class weights were also calculated for weighted loss functions. The validation and test loaders preserved a fixed sample order to ensure consistent and reproducible model evaluation.


```python
def filter_excluded_paths(dataset, excluded_paths=None, split_name=""):
    excluded_paths = set(excluded_paths or [])
    if not excluded_paths:
        return dataset

    original_n = len(dataset.samples)
    kept_samples = [
        (path, target)
        for path, target in dataset.samples
        if canonical_path(path) not in excluded_paths
    ]

    if not kept_samples:
        raise RuntimeError(f"All samples were excluded from split '{split_name}'. Check duplicate filtering.")

    dataset.samples = kept_samples
    dataset.imgs = kept_samples
    dataset.targets = [target for _, target in kept_samples]

    removed_n = original_n - len(kept_samples)
    if removed_n:
        print(f"Excluded {removed_n} leakage-risk duplicate image(s) from {split_name}.")
    return dataset

def limit_dataset_per_class(dataset, max_samples_per_class=None):
    if max_samples_per_class is None:
        return dataset
    rng = np.random.default_rng(CONFIG["seed"])
    targets = np.array(dataset.targets)
    selected_indices = []
    for cls_idx in range(len(dataset.classes)):
        cls_indices = np.where(targets == cls_idx)[0]
        rng.shuffle(cls_indices)
        selected_indices.extend(cls_indices[:max_samples_per_class].tolist())
    rng.shuffle(selected_indices)
    subset = Subset(dataset, selected_indices)
    subset.classes = dataset.classes
    subset.class_to_idx = dataset.class_to_idx
    return subset

def get_targets(dataset):
    if isinstance(dataset, Subset):
        return np.array(dataset.dataset.targets)[dataset.indices]
    return np.array(dataset.targets)

def assert_class_coverage(dataset, split_name):
    targets = get_targets(dataset)
    missing = [
        dataset.classes[i]
        for i in range(len(dataset.classes))
        if not np.any(targets == i)
    ]
    if missing:
        raise RuntimeError(f"Split '{split_name}' has no samples for class(es): {missing}")

def make_loaders(train_transform, batch_size=None, max_samples_per_class=None, use_weighted_sampler=True):
    batch_size = batch_size or CONFIG["batch_size"]
    excluded_paths = globals().get("EXCLUDED_IMAGE_PATHS", set())

    train_ds = datasets.ImageFolder(DATA_ROOT / "train", transform=train_transform)
    val_ds = datasets.ImageFolder(DATA_ROOT / "val", transform=eval_transform)
    test_ds = datasets.ImageFolder(DATA_ROOT / "test", transform=eval_transform)

    if not (train_ds.classes == val_ds.classes == test_ds.classes):
        raise RuntimeError("Class folders are not identical across train/val/test splits.")

    train_ds = filter_excluded_paths(train_ds, excluded_paths, split_name="train")
    val_ds = filter_excluded_paths(val_ds, excluded_paths, split_name="val")
    test_ds = filter_excluded_paths(test_ds, excluded_paths, split_name="test")

    train_ds = limit_dataset_per_class(train_ds, max_samples_per_class)
    val_ds = limit_dataset_per_class(val_ds, max_samples_per_class)
    test_ds = limit_dataset_per_class(test_ds, max_samples_per_class)

    assert_class_coverage(train_ds, "train")
    assert_class_coverage(val_ds, "val")
    assert_class_coverage(test_ds, "test")

    class_names_local = train_ds.classes
    targets = get_targets(train_ds)
    class_counts = np.bincount(targets, minlength=len(class_names_local))

    sampler = None
    shuffle = True
    if use_weighted_sampler:
        class_sample_weights = len(targets) / np.maximum(class_counts, 1)
        sample_weights = class_sample_weights[targets]
        sampler = WeightedRandomSampler(
            weights=torch.as_tensor(sample_weights, dtype=torch.double),
            num_samples=len(sample_weights),
            replacement=True,
        )
        shuffle = False

    loader_kwargs = {
        "batch_size": batch_size,
        "num_workers": CONFIG["num_workers"],
        "pin_memory": device.type == "cuda",
    }
    if CONFIG["num_workers"] > 0:
        loader_kwargs["persistent_workers"] = True

    train_loader = DataLoader(train_ds, shuffle=shuffle, sampler=sampler, **loader_kwargs)
    val_loader = DataLoader(val_ds, shuffle=False, **loader_kwargs)
    test_loader = DataLoader(test_ds, shuffle=False, **loader_kwargs)

    ce_weights = len(targets) / np.maximum(class_counts, 1)
    ce_weights = ce_weights / ce_weights.mean()
    ce_weights = torch.tensor(ce_weights, dtype=torch.float32, device=device)

    return train_loader, val_loader, test_loader, class_names_local, ce_weights

train_loader_base, val_loader, test_loader, class_names, class_weights = make_loaders(
    train_base_transform,
    max_samples_per_class=CONFIG["max_samples_per_class"],
    use_weighted_sampler=CONFIG["use_weighted_sampler"],
)
NUM_CLASSES = len(class_names)
print("Class names:", class_names)
print("Class weights:", class_weights.detach().cpu().numpy().round(3))
print("Train batches:", len(train_loader_base), "Val batches:", len(val_loader), "Test batches:", len(test_loader))
print("Leakage status:", globals().get("DATASET_LEAKAGE_STATUS", "Duplicate scan was not run."))
```

# 6. Training, evaluation, and efficiency utilities

This section defined reusable utilities for model training, validation, testing, checkpoint management, and efficiency analysis. These functions ensured that all experimental approaches were trained and evaluated using the same procedures and metrics.

This cell defined the complete training and evaluation pipeline. It included weighted focal loss and cross-entropy loss, label smoothing, MixUp, mixed-precision training, gradient clipping, learning-rate scheduling, early stopping, validation-based best-checkpoint selection, TTA prediction averaging, classification metrics, confusion matrices, and training-history plots.

These utilities were used to train each model consistently and to select the best model state according to validation Macro F1-score. Frozen Batch Normalization layers were kept in evaluation mode during partial fine-tuning to avoid unintended changes in their stored statistics.


```python
def clear_memory():
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

class WeightedFocalLoss(nn.Module):
    def __init__(self, weight=None, gamma=1.0, label_smoothing=0.0):
        super().__init__()
        self.gamma = float(gamma)
        self.label_smoothing = float(label_smoothing)
        if weight is not None:
            self.register_buffer("weight", weight.detach().clone())
        else:
            self.weight = None

    def forward(self, logits, targets):
        ce_loss = F.cross_entropy(
            logits,
            targets,
            weight=self.weight,
            label_smoothing=self.label_smoothing,
            reduction="none",
        )
        pt = torch.softmax(logits, dim=1).gather(1, targets.unsqueeze(1)).squeeze(1).clamp_min(1e-6)
        focal_factor = (1.0 - pt).pow(self.gamma)
        return (focal_factor * ce_loss).mean()


def make_criterion(class_weights, label_smoothing=None, loss_name="cross_entropy", focal_gamma=None):
    if label_smoothing is None:
        label_smoothing = CONFIG["label_smoothing"]

    loss_name = str(loss_name).lower()
    if loss_name in {"focal", "weighted_focal", "focal_loss"}:
        gamma = CONFIG.get("baseline_focal_gamma", 1.25) if focal_gamma is None else focal_gamma
        return WeightedFocalLoss(
            weight=class_weights,
            gamma=gamma,
            label_smoothing=label_smoothing,
        )

    return nn.CrossEntropyLoss(
        weight=class_weights,
        label_smoothing=label_smoothing,
    )

def compute_metrics(y_true, y_prob, class_names):
    y_pred = np.argmax(y_prob, axis=1)
    metrics = {
        "accuracy": accuracy_score(y_true, y_pred),
        "balanced_accuracy": balanced_accuracy_score(y_true, y_pred),
        "macro_f1": f1_score(y_true, y_pred, average="macro", zero_division=0),
        "weighted_f1": f1_score(y_true, y_pred, average="weighted", zero_division=0),
        "macro_precision": precision_score(y_true, y_pred, average="macro", zero_division=0),
        "macro_recall": recall_score(y_true, y_pred, average="macro", zero_division=0),
    }
    try:
        metrics["roc_auc_ovr_macro"] = roc_auc_score(y_true, y_prob, multi_class="ovr", average="macro")
    except Exception:
        metrics["roc_auc_ovr_macro"] = np.nan
    return metrics

@torch.no_grad()
def collect_predictions(model, loader):
    model.eval()
    all_probs = []
    all_targets = []
    for images, targets in tqdm(loader, desc="Predict", leave=False):
        images = images.to(device, non_blocking=True)
        logits = model(images)
        probs = torch.softmax(logits, dim=1).detach().cpu().numpy()
        all_probs.append(probs)
        all_targets.append(targets.numpy())
    return np.concatenate(all_targets), np.concatenate(all_probs)

def plot_confusion(y_true, y_prob, class_names, title):
    y_pred = np.argmax(y_prob, axis=1)
    cm = confusion_matrix(y_true, y_pred)
    denom = np.clip(cm.sum(axis=1, keepdims=True), 1, None)
    cm_norm = cm / denom

    fig, axes = plt.subplots(1, 2, figsize=(15, 5))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=class_names, yticklabels=class_names, ax=axes[0])
    axes[0].set_title(f"{title}: counts")
    axes[0].set_xlabel("Predicted")
    axes[0].set_ylabel("True")

    sns.heatmap(cm_norm, annot=True, fmt=".2f", cmap="Greens", xticklabels=class_names, yticklabels=class_names, ax=axes[1])
    axes[1].set_title(f"{title}: normalized")
    axes[1].set_xlabel("Predicted")
    axes[1].set_ylabel("True")
    plt.tight_layout()
    plt.show()

def evaluate_model(model, loader, class_names, title, show_report=True, show_cm=True):
    y_true, y_prob = collect_predictions(model, loader)
    metrics = compute_metrics(y_true, y_prob, class_names)
    print(f"\n{title}")
    for key, value in metrics.items():
        print(f"{key:>22}: {value:.4f}" if pd.notna(value) else f"{key:>22}: NaN")
    if show_report:
        print("\nClassification report")
        print(classification_report(y_true, np.argmax(y_prob, axis=1), target_names=class_names, digits=4, zero_division=0))
    if show_cm:
        plot_confusion(y_true, y_prob, class_names, title)
    return metrics, y_true, y_prob

def make_eval_loader(split, transform, batch_size=None, max_samples_per_class=None):
    batch_size = batch_size or CONFIG["batch_size"]
    excluded_paths = globals().get("EXCLUDED_IMAGE_PATHS", set())
    ds = datasets.ImageFolder(DATA_ROOT / split, transform=transform)
    ds = filter_excluded_paths(ds, excluded_paths, split_name=split)
    ds = limit_dataset_per_class(ds, max_samples_per_class)
    assert_class_coverage(ds, split)

    loader_kwargs = {
        "batch_size": batch_size,
        "num_workers": CONFIG["num_workers"],
        "pin_memory": device.type == "cuda",
    }
    if CONFIG["num_workers"] > 0:
        loader_kwargs["persistent_workers"] = True
    return DataLoader(ds, shuffle=False, **loader_kwargs), ds.classes

def baseline_tta_transforms():
    return [
        eval_transform,
        transforms.Compose([
            transforms.Resize((236, 236)),
            transforms.Grayscale(num_output_channels=3),
            transforms.CenterCrop(IMG_SIZE),
            transforms.ToTensor(),
            transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
        ]),
        transforms.Compose([
            transforms.Resize((244, 244)),
            transforms.Grayscale(num_output_channels=3),
            transforms.CenterCrop(IMG_SIZE),
            transforms.ToTensor(),
            transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
        ]),
    ]

@torch.no_grad()
def collect_predictions_tta(model, split="test", transforms_list=None, batch_size=None):
    transforms_list = transforms_list or baseline_tta_transforms()
    y_true_ref = None
    all_probs = []
    for idx, transform in enumerate(transforms_list, start=1):
        loader, local_class_names = make_eval_loader(
            split,
            transform,
            batch_size=batch_size,
            max_samples_per_class=CONFIG["max_samples_per_class"],
        )
        if list(local_class_names) != list(class_names):
            raise RuntimeError("Class order mismatch during TTA evaluation.")
        y_true, y_prob = collect_predictions(model, loader)
        if y_true_ref is None:
            y_true_ref = y_true
        elif not np.array_equal(y_true_ref, y_true):
            raise RuntimeError("Sample order mismatch during TTA evaluation.")
        all_probs.append(y_prob)
        print(f"TTA pass {idx}/{len(transforms_list)} complete for {split} split.")
    return y_true_ref, np.mean(all_probs, axis=0)

def evaluate_model_tta(model, split, class_names, title, show_report=True, show_cm=True):
    y_true, y_prob = collect_predictions_tta(model, split=split)
    metrics = compute_metrics(y_true, y_prob, class_names)
    print(f"\n{title}")
    for key, value in metrics.items():
        print(f"{key:>22}: {value:.4f}" if pd.notna(value) else f"{key:>22}: NaN")
    if show_report:
        print("\nClassification report")
        print(classification_report(y_true, np.argmax(y_prob, axis=1), target_names=class_names, digits=4, zero_division=0))
    if show_cm:
        plot_confusion(y_true, y_prob, class_names, title)
    return metrics, y_true, y_prob

def set_frozen_batchnorm_eval(model):
    """Keep frozen BatchNorm layers in eval mode while trainable layers keep learning."""
    for module in model.modules():
        if isinstance(module, nn.modules.batchnorm._BatchNorm):
            direct_params = list(module.parameters(recurse=False))
            if direct_params and not any(p.requires_grad for p in direct_params):
                module.eval()

def apply_mixup(images, targets, alpha):
    if alpha is None or alpha <= 0 or images.size(0) < 2:
        return images, targets, None, 1.0
    lam = np.random.beta(alpha, alpha)
    index = torch.randperm(images.size(0), device=images.device)
    mixed_images = lam * images + (1.0 - lam) * images[index]
    return mixed_images, targets, targets[index], float(lam)


def mixup_loss(criterion, logits, targets_a, targets_b=None, lam=1.0):
    if targets_b is None:
        return criterion(logits, targets_a)
    return lam * criterion(logits, targets_a) + (1.0 - lam) * criterion(logits, targets_b)


def train_one_epoch(model, loader, criterion, optimizer, scaler, mixup_alpha=0.0, mixup_prob=0.0):
    model.train()
    set_frozen_batchnorm_eval(model)
    running_loss = 0.0
    correct = 0
    total = 0

    for images, targets in tqdm(loader, desc="Train", leave=False):
        images = images.to(device, non_blocking=True)
        targets = targets.to(device, non_blocking=True)
        original_targets = targets

        use_mixup = mixup_alpha and mixup_alpha > 0 and random.random() < mixup_prob
        if use_mixup:
            images, targets_a, targets_b, lam = apply_mixup(images, targets, mixup_alpha)
        else:
            targets_a, targets_b, lam = targets, None, 1.0

        optimizer.zero_grad(set_to_none=True)
        use_amp = CONFIG["use_amp"] and device.type == "cuda"
        with autocast(enabled=use_amp):
            logits = model(images)
            loss = mixup_loss(criterion, logits, targets_a, targets_b, lam)

        if use_amp:
            scaler.scale(loss).backward()
            if CONFIG.get("grad_clip_norm"):
                scaler.unscale_(optimizer)
                torch.nn.utils.clip_grad_norm_(model.parameters(), CONFIG["grad_clip_norm"])
            scaler.step(optimizer)
            scaler.update()
        else:
            loss.backward()
            if CONFIG.get("grad_clip_norm"):
                torch.nn.utils.clip_grad_norm_(model.parameters(), CONFIG["grad_clip_norm"])
            optimizer.step()

        batch_size = original_targets.size(0)
        running_loss += loss.item() * batch_size
        correct += (logits.argmax(dim=1) == original_targets).sum().item()
        total += batch_size

    return running_loss / max(total, 1), correct / max(total, 1)

def copy_state_dict_to_cpu(model):
    return {
        name: tensor.detach().cpu().clone()
        for name, tensor in model.state_dict().items()
    }

def fit_model(model, train_loader, val_loader, criterion, optimizer, epochs, monitor="macro_f1", phase_name="", checkpoint_tag=None, early_stopping_patience=None, min_epochs_before_stopping=None, scheduler_patience=None, mixup_alpha=0.0, mixup_prob=0.0):
    scaler = GradScaler(enabled=CONFIG["use_amp"] and device.type == "cuda")
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode="max",
        factor=0.35,
        patience=2 if scheduler_patience is None else scheduler_patience,
    )

    history = []
    best_score = -np.inf
    best_epoch = -1
    patience_counter = 0
    best_state = None
    if min_epochs_before_stopping is None:
        min_epochs_before_stopping = CONFIG.get("min_epochs_before_stopping", 0)
    if early_stopping_patience is None:
        early_stopping_patience = CONFIG["early_stopping_patience"]
    min_epochs = min(min_epochs_before_stopping, epochs)

    for epoch in range(1, epochs + 1):
        train_loss, train_acc = train_one_epoch(
            model,
            train_loader,
            criterion,
            optimizer,
            scaler,
            mixup_alpha=mixup_alpha,
            mixup_prob=mixup_prob,
        )
        y_val, p_val = collect_predictions(model, val_loader)
        val_metrics = compute_metrics(y_val, p_val, class_names)
        scheduler.step(val_metrics[monitor])

        row = {
            "epoch": epoch,
            "stage": phase_name,
            "train_loss": train_loss,
            "train_accuracy": train_acc,
            **{f"val_{k}": v for k, v in val_metrics.items()},
            "lr_backbone_or_main": optimizer.param_groups[0]["lr"],
            "lr_head": optimizer.param_groups[-1]["lr"],
        }
        history.append(row)

        prefix = f"[{phase_name}] " if phase_name else ""
        print(
            f"{prefix}Epoch {epoch:02d}/{epochs} | "
            f"loss={train_loss:.4f} | train_acc={train_acc:.4f} | "
            f"val_acc={val_metrics['accuracy']:.4f} | val_macro_f1={val_metrics['macro_f1']:.4f}"
        )

        is_best_epoch = False
        if val_metrics[monitor] > best_score:
            best_score = val_metrics[monitor]
            best_epoch = epoch
            patience_counter = 0
            best_state = copy_state_dict_to_cpu(model)
            is_best_epoch = True
        else:
            patience_counter += 1

        autosave_fn = globals().get("autosave_training_checkpoint")
        checkpoint_name = checkpoint_tag or phase_name or model.__class__.__name__
        should_checkpoint = (
            callable(autosave_fn)
            and CONFIG.get("enable_checkpointing", True)
            and epoch % max(int(CONFIG.get("checkpoint_every_n_epochs", 1)), 1) == 0
        )
        if should_checkpoint:
            autosave_fn(
                model=model,
                optimizer=optimizer,
                scaler=scaler,
                scheduler=scheduler,
                history_df=pd.DataFrame(history),
                checkpoint_tag=checkpoint_name,
                epoch=epoch,
                monitor=monitor,
                best_state=best_state,
                best_score=best_score,
                best_epoch=best_epoch,
                is_best=is_best_epoch,
            )

        if not is_best_epoch and epoch >= min_epochs and patience_counter >= early_stopping_patience:
            print(f"Early stopping at epoch {epoch}. Best epoch: {best_epoch}, best {monitor}: {best_score:.4f}")
            break

    model.load_state_dict(best_state)
    training_state = {
        "model_state": best_state,
        "best_epoch": best_epoch,
        "best_score": best_score,
        "class_names": class_names,
        "config": CONFIG,
    }
    return model, pd.DataFrame(history), training_state

def merge_histories(*histories):
    merged = []
    epoch_offset = 0
    for hist in histories:
        if hist is None or hist.empty:
            continue
        h = hist.copy()
        h["epoch"] = h["epoch"] + epoch_offset
        epoch_offset = int(h["epoch"].max())
        merged.append(h)
    return pd.concat(merged, ignore_index=True) if merged else pd.DataFrame()

def choose_better_state(model, current_state, candidate_state):
    if candidate_state["best_score"] >= current_state["best_score"]:
        model.load_state_dict(candidate_state["model_state"])
        return candidate_state
    model.load_state_dict(current_state["model_state"])
    return current_state

def plot_history(history_df, title):
    if history_df.empty:
        return
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    axes[0].plot(history_df["epoch"], history_df["train_loss"], marker="o")
    axes[0].set_title(f"{title}: training loss")
    axes[0].set_xlabel("Epoch")
    axes[0].set_ylabel("Loss")

    axes[1].plot(history_df["epoch"], history_df["train_accuracy"], label="train acc", marker="o")
    axes[1].plot(history_df["epoch"], history_df["val_accuracy"], label="val acc", marker="o")
    axes[1].plot(history_df["epoch"], history_df["val_macro_f1"], label="val macro F1", marker="o")
    axes[1].set_title(f"{title}: validation metrics")
    axes[1].set_xlabel("Epoch")
    axes[1].legend()
    plt.tight_layout()
    plt.show()
```

This cell defined functions for measuring model efficiency, including parameter count, trainable parameters, checkpoint size, compressed size, global sparsity, and inference latency. It also created a common experiment registry and results table for comparing predictive performance with computational cost.

Checkpoint utilities were used to automatically save the latest and best model states, restore previous experiments, replay saved training histories, and avoid unnecessary retraining after interrupted sessions. Final model selection and ensemble candidate ranking were based on validation Macro F1-score, with model size and latency used as secondary criteria.


```python
def count_parameters(model, trainable_only=False):
    params = model.parameters()
    if trainable_only:
        params = [p for p in model.parameters() if p.requires_grad]
    return int(sum(p.numel() for p in params))

def state_dict_sizes_mb(model):
    buffer = io.BytesIO()
    torch.save(model.state_dict(), buffer)
    raw = buffer.getvalue()
    compressed = gzip.compress(raw)
    return len(raw) / (1024 ** 2), len(compressed) / (1024 ** 2)

def global_sparsity(model):
    zero = 0
    total = 0
    for name, p in model.named_parameters():
        if "weight" in name and p.ndim > 1:
            data = p.detach()
            zero += int(torch.sum(data == 0).item())
            total += data.numel()
    return zero / max(total, 1)

@torch.no_grad()
def measure_latency_ms(model, batch_size=1):
    model.eval()
    sample = torch.randn(batch_size, 3, IMG_SIZE, IMG_SIZE, device=device)
    runs = CONFIG["latency_runs"]
    warmup = CONFIG["latency_warmup"]

    for _ in range(warmup):
        _ = model(sample)
    if device.type == "cuda":
        torch.cuda.synchronize()
    start = time.perf_counter()
    for _ in range(runs):
        _ = model(sample)
    if device.type == "cuda":
        torch.cuda.synchronize()
    elapsed = time.perf_counter() - start
    return (elapsed / runs) * 1000.0 / batch_size

def model_efficiency(model):
    raw_mb, gz_mb = state_dict_sizes_mb(model)
    return {
        "params_m": count_parameters(model) / 1e6,
        "trainable_params_m": count_parameters(model, trainable_only=True) / 1e6,
        "state_dict_mb": raw_mb,
        "gz_state_dict_mb": gz_mb,
        "sparsity": global_sparsity(model),
        "latency_ms_img": measure_latency_ms(model),
    }

experiment_results = []
model_state_registry = {}

def resolve_checkpoint_dir():
    env_dir = os.environ.get("CHESTX6_CHECKPOINT_DIR")
    if env_dir:
        return Path(env_dir).expanduser()

    home = Path.home()
    if os.name == "nt":
        desktop_candidates = [
            home / "OneDrive" / "Pulpit",
            home / "OneDrive" / "Desktop",
            home / "Desktop",
            home / "Pulpit",
        ]
        for desktop_dir in desktop_candidates:
            if desktop_dir.exists():
                return desktop_dir / "DATA"
        if (home / "OneDrive").exists():
            return home / "OneDrive" / "Pulpit" / "DATA"
        return home / "Desktop" / "DATA"

    if Path("/content").exists():
        return Path("/content") / "DATA"

    for desktop_dir in [home / "Desktop", home / "Pulpit"]:
        if desktop_dir.exists():
            return desktop_dir / "DATA"
    return Path.cwd() / "DATA"

def safe_checkpoint_name(name):
    allowed = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789-_."
    safe = "".join(ch if ch in allowed else "_" for ch in str(name))
    safe = "_".join(part for part in safe.split("_") if part)
    return safe[:120] or "checkpoint"

SAVE_DIR = resolve_checkpoint_dir()
SAVE_DIR.mkdir(parents=True, exist_ok=True)

MODELS_DIR = SAVE_DIR
RESULTS_CSV = SAVE_DIR / "results.csv"

print(f"Checkpoint directory: {SAVE_DIR}")
print("This notebook writes only .pt checkpoint files and results.csv to the checkpoint directory.")

def save_results_csv():
    df = results_table()
    df.to_csv(RESULTS_CSV, index=False)
    print(f"Saved results to: {RESULTS_CSV}")

def autosave_training_checkpoint(
    model,
    optimizer,
    scaler,
    scheduler,
    history_df,
    checkpoint_tag,
    epoch,
    monitor,
    best_state,
    best_score,
    best_epoch,
    is_best=False,
):
    if not CONFIG.get("enable_checkpointing", True):
        return

    tag = safe_checkpoint_name(checkpoint_tag)
    payload = {
        "checkpoint_type": "autosave",
        "checkpoint_tag": checkpoint_tag,
        "epoch": epoch,
        "monitor": monitor,
        "best_epoch": best_epoch,
        "best_score": float(best_score),
        "model_state_dict": copy_state_dict_to_cpu(model),
        "best_model_state_dict": best_state,
        "optimizer_state_dict": optimizer.state_dict() if optimizer is not None else None,
        "scaler_state_dict": scaler.state_dict() if scaler is not None else None,
        "scheduler_state_dict": scheduler.state_dict() if scheduler is not None else None,
        "history": history_df.to_dict("records") if history_df is not None else None,
        "config": CONFIG,
        "class_names": globals().get("class_names"),
    }

    latest_path = SAVE_DIR / f"{tag}_latest.pt"
    torch.save(payload, latest_path)

    if is_best:
        best_path = SAVE_DIR / f"{tag}_best.pt"
        torch.save(payload, best_path)
        print(f"Autosaved best checkpoint: {best_path}")

def save_model_checkpoint(state_key, model, training_state, metrics=None, history_df=None):
    path = MODELS_DIR / f"{safe_checkpoint_name(state_key)}.pt"

    torch.save({
        "checkpoint_type": "final_experiment",
        "state_key": state_key,
        "model_family": model_state_registry.get(state_key, {}).get("model_family"),
        "model_state_dict": model.state_dict(),
        "training_state": training_state,
        "metrics": metrics,
        "history": history_df.to_dict("records") if history_df is not None else None,
        "config": CONFIG,
        "class_names": class_names,
    }, path)

    print(f"Saved model checkpoint to: {path}")

def list_saved_checkpoints():
    checkpoints = sorted(SAVE_DIR.glob("*.pt"), key=lambda p: p.stat().st_mtime, reverse=True)
    return pd.DataFrame({
        "path": [str(p) for p in checkpoints],
        "size_mb": [p.stat().st_size / (1024 ** 2) for p in checkpoints],
        "modified": [pd.Timestamp.fromtimestamp(p.stat().st_mtime) for p in checkpoints],
    })

def load_saved_checkpoint(path):
    path = Path(path)
    try:
        checkpoint = torch.load(path, map_location=device, weights_only=False)
    except TypeError:
        checkpoint = torch.load(path, map_location=device)
    print(f"Loaded checkpoint: {path}")
    print("Checkpoint type:", checkpoint.get("checkpoint_type", "unknown"))
    return checkpoint


def final_checkpoint_path(state_key):
    return MODELS_DIR / f"{safe_checkpoint_name(state_key)}.pt"

def find_existing_checkpoint(state_key, extra_filenames=None):
    candidates = []
    for filename in extra_filenames or []:
        if filename:
            candidates.append(SAVE_DIR / filename)

    tag = safe_checkpoint_name(state_key)
    candidates.extend([
        final_checkpoint_path(state_key),
        SAVE_DIR / f"{tag}_best.pt",
        SAVE_DIR / f"{tag}_latest.pt",
    ])

    seen = set()
    for path in candidates:
        path = Path(path)
        if path in seen:
            continue
        seen.add(path)
        if path.exists():
            return path
    return None

def checkpoint_model_state_dict(checkpoint):
    for key in ["model_state_dict", "best_model_state_dict"]:
        state_dict = checkpoint.get(key)
        if state_dict is not None:
            return state_dict
    training_state = checkpoint.get("training_state") or {}
    return training_state.get("model_state")

def checkpoint_history_df(checkpoint, stage_rename=None):
    history = checkpoint.get("history")
    if history is None:
        history = (checkpoint.get("training_state") or {}).get("history")
    history_df = pd.DataFrame(history or [])

    if not history_df.empty:
        if "stage" not in history_df.columns:
            history_df["stage"] = "checkpoint replay"
        if stage_rename:
            history_df["stage"] = history_df["stage"].replace(stage_rename)
    return history_df

def checkpoint_training_state(checkpoint, model, history_df=None):
    saved_training_state = checkpoint.get("training_state") or {}
    best_epoch = saved_training_state.get("best_epoch", checkpoint.get("best_epoch"))
    best_score = saved_training_state.get("best_score", checkpoint.get("best_score"))

    if (best_score is None or pd.isna(best_score)) and history_df is not None and not history_df.empty:
        if "val_macro_f1" in history_df:
            best_score = float(history_df["val_macro_f1"].max())
            best_epoch = int(history_df["val_macro_f1"].idxmax()) + 1
    if best_score is None or pd.isna(best_score):
        best_score = np.nan

    return {
        "model_state": copy_state_dict_to_cpu(model),
        "best_epoch": best_epoch,
        "best_score": float(best_score) if not pd.isna(best_score) else np.nan,
        "class_names": class_names,
        "config": CONFIG,
    }

def display_model_profile(model):
    if summary is None:
        print(model)
        return

    was_training = model.training
    model.eval()
    display(summary(model, input_size=(1, 3, IMG_SIZE, IMG_SIZE), verbose=0))
    if was_training:
        model.train()

def drop_experiment_result(approach_name):
    experiment_results[:] = [
        row for row in experiment_results
        if row.get("approach") != approach_name
    ]

def replay_checkpoint_history(history_df, stage_total_epochs=None, checkpoint_tags=None, default_checkpoint_tag=None):
    if not CONFIG.get("replay_checkpoint_history", True):
        return
    if history_df is None or history_df.empty:
        print("Checkpoint has no saved epoch history to replay.")
        return

    stage_total_epochs = stage_total_epochs or {}
    checkpoint_tags = checkpoint_tags or {}
    work_df = history_df.copy()
    if "stage" not in work_df.columns:
        work_df["stage"] = "checkpoint replay"

    for stage, stage_df in work_df.groupby("stage", sort=False):
        total_epochs = int(stage_total_epochs.get(stage, len(stage_df)))
        best_score = -np.inf
        best_epoch = None

        for local_epoch, (_, row) in enumerate(stage_df.iterrows(), start=1):
            train_loss = float(row.get("train_loss", np.nan))
            train_acc = float(row.get("train_accuracy", np.nan))
            val_acc = float(row.get("val_accuracy", np.nan))
            val_macro_f1 = float(row.get("val_macro_f1", np.nan))

            print(
                f"[{stage}] Epoch {local_epoch:02d}/{total_epochs} | "
                f"loss={train_loss:.4f} | train_acc={train_acc:.4f} | "
                f"val_acc={val_acc:.4f} | val_macro_f1={val_macro_f1:.4f}"
            )

            if pd.notna(val_macro_f1) and val_macro_f1 > best_score:
                best_score = val_macro_f1
                best_epoch = local_epoch
                tag = checkpoint_tags.get(stage, default_checkpoint_tag)
                if tag:
                    print(f"Autosaved best checkpoint: {SAVE_DIR / (safe_checkpoint_name(tag) + '_best.pt')}")

        if total_epochs and len(stage_df) < total_epochs and best_epoch is not None:
            print(
                f"Early stopping at epoch {len(stage_df)}. "
                f"Best epoch: {best_epoch}, best macro_f1: {best_score:.4f}"
            )

def load_checkpoint_experiment(
    checkpoint_path,
    model,
    state_key,
    model_family,
    approach_name,
    history_title,
    eval_title,
    eval_loader=None,
    eval_tta=False,
    tta_split="test",
    notes="",
    stage_rename=None,
    stage_total_epochs=None,
    checkpoint_tags=None,
    default_checkpoint_tag=None,
):
    print(f"Loading existing checkpoint instead of retraining: {checkpoint_path}")
    checkpoint = load_saved_checkpoint(checkpoint_path)
    state_dict = checkpoint_model_state_dict(checkpoint)
    if state_dict is None:
        raise RuntimeError(f"No model weights found in checkpoint: {checkpoint_path}")

    model.load_state_dict(state_dict)
    model.eval()
    display_model_profile(model)

    history_df = checkpoint_history_df(checkpoint, stage_rename=stage_rename)
    replay_checkpoint_history(
        history_df,
        stage_total_epochs=stage_total_epochs,
        checkpoint_tags=checkpoint_tags,
        default_checkpoint_tag=default_checkpoint_tag or state_key,
    )
    plot_history(history_df, history_title)

    saved_metrics = checkpoint.get("metrics")
    if CONFIG.get("reevaluate_loaded_checkpoints", True) or saved_metrics is None:
        if eval_tta:
            metrics, _, _ = evaluate_model_tta(model, tta_split, class_names, eval_title)
        else:
            metrics, _, _ = evaluate_model(model, eval_loader or test_loader, class_names, eval_title)
    else:
        metrics = saved_metrics
        print(f"Using metrics saved in checkpoint for: {eval_title}")

    training_state = checkpoint_training_state(checkpoint, model, history_df)
    drop_experiment_result(approach_name)
    register_model_state(state_key, training_state, model_family)
    add_experiment_result(
        approach_name,
        model,
        metrics,
        history_df,
        state_key,
        model_family=model_family,
        notes=notes,
        best_val_score=training_state["best_score"],
    )
    return model, history_df, training_state, metrics

def register_model_state(state_key, training_state, model_family):
    model_state_registry[state_key] = {
        "model_family": model_family,
        "model_state": training_state["model_state"],
        "best_epoch": training_state.get("best_epoch"),
        "best_score": training_state.get("best_score"),
    }
    return state_key

def training_state_from_registry(state_key):
    entry = model_state_registry[state_key]
    best_score = entry.get("best_score")

    if best_score is None or pd.isna(best_score):
        best_score = -np.inf

    return {
        "model_state": copy.deepcopy(entry["model_state"]),
        "best_epoch": entry.get("best_epoch"),
        "best_score": float(best_score),
        "class_names": class_names,
        "config": CONFIG,
    }

def add_experiment_result(approach, model, metrics, history_df=None, state_key="", model_family=None, notes="", best_val_score=None):
    eff = model_efficiency(model)
    if best_val_score is None:
        best_val_score = float(history_df["val_macro_f1"].max()) if history_df is not None and not history_df.empty else np.nan
    else:
        best_val_score = float(best_val_score)
    row = {
        "approach": approach,
        "model_family": model_family or approach,
        "state_key": state_key,
        "best_val_macro_f1": best_val_score,
        **metrics,
        **eff,
        "notes": notes,
    }
    experiment_results.append(row)
    return row

def results_table():
    if not experiment_results:
        return pd.DataFrame()
    cols = [
        "approach", "model_family",
        "accuracy", "balanced_accuracy", "macro_precision", "macro_recall", "macro_f1", "weighted_f1", "roc_auc_ovr_macro",
        "best_val_macro_f1", "params_m", "trainable_params_m", "state_dict_mb", "gz_state_dict_mb",
        "sparsity", "latency_ms_img", "state_key", "notes"
    ]
    df = pd.DataFrame(experiment_results)
    return df[[c for c in cols if c in df.columns]].sort_values(["macro_f1", "accuracy"], ascending=False)

def sort_for_validation_selection(df, top_k=None):
    if df.empty:
        return df.copy()
    selection_df = df.dropna(subset=["best_val_macro_f1"]).copy()
    if selection_df.empty:
        raise RuntimeError("No validation scores are available for model selection.")

    sort_cols = ["best_val_macro_f1"]
    ascending = [False]
    for col in ["params_m", "latency_ms_img", "approach"]:
        if col in selection_df.columns:
            sort_cols.append(col)
            ascending.append(True)

    selection_df = selection_df.sort_values(sort_cols, ascending=ascending, na_position="last")
    if top_k is not None:
        selection_df = selection_df.head(top_k)
    return selection_df
```

# 7. Model definitions

This section defined all CNN architectures used in the experiments, including the custom baseline models and the ImageNet-pretrained transfer-learning models. It also provided utilities for freezing, unfreezing, fine-tuning, and reconstructing the selected architectures.

The model definitions were prepared in the following steps:

1. `ConvBNAct` combined convolution, Batch Normalization, and SiLU activation into a reusable feature-extraction block.
2. `SqueezeExcite` introduced channel attention by learning which feature channels should receive greater importance.
3. `ResidualSEBlock` combined residual connections, Squeeze-and-Excitation attention, spatial downsampling, and dropout.
4. `SimpleCNN` was defined as the main scratch baseline, using four convolutional stages, Batch Normalization, ReLU activations, pooling, global average pooling, and a dropout-based classifier.
5. `OptimizedBaselineCNN` was created as a more advanced scratch model using residual SE blocks, combined global average and maximum pooling, and a deeper regularized classification head.
6. Transfer-learning models were created from pretrained ResNet50, DenseNet121, EfficientNet-B0, and MobileNetV2 architectures. Their original classification layers were replaced with a common multi-layer head adapted to the six output classes.
7. The pretrained backbones could initially be frozen so that only the new classification heads were trained.
8. Selected final or deeper backbone blocks could later be unfrozen for staged fine-tuning.
9. A discriminative AdamW optimizer assigned smaller learning rates to pretrained backbone parameters and larger learning rates to newly added classifier parameters.
10. A common model-building function allowed each saved architecture to be reconstructed from its registered model family.


```python
class ConvBNAct(nn.Sequential):
    def __init__(self, in_channels, out_channels, kernel_size=3, stride=1):
        padding = kernel_size // 2
        super().__init__(
            nn.Conv2d(in_channels, out_channels, kernel_size=kernel_size, stride=stride, padding=padding, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.SiLU(inplace=True),
        )


class SqueezeExcite(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()
        hidden = max(channels // reduction, 8)
        self.pool = nn.AdaptiveAvgPool2d(1)
        self.fc = nn.Sequential(
            nn.Conv2d(channels, hidden, kernel_size=1),
            nn.SiLU(inplace=True),
            nn.Conv2d(hidden, channels, kernel_size=1),
            nn.Sigmoid(),
        )

    def forward(self, x):
        return x * self.fc(self.pool(x))


class ResidualSEBlock(nn.Module):
    def __init__(self, in_channels, out_channels, stride=1, dropout=0.0):
        super().__init__()
        self.conv1 = ConvBNAct(in_channels, out_channels, stride=stride)
        self.conv2 = nn.Sequential(
            nn.Conv2d(out_channels, out_channels, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(out_channels),
        )
        self.se = SqueezeExcite(out_channels)
        self.dropout = nn.Dropout2d(dropout) if dropout and dropout > 0 else nn.Identity()
        if stride != 1 or in_channels != out_channels:
            self.shortcut = nn.Sequential(
                nn.Conv2d(in_channels, out_channels, kernel_size=1, stride=stride, bias=False),
                nn.BatchNorm2d(out_channels),
            )
        else:
            self.shortcut = nn.Identity()
        self.act = nn.SiLU(inplace=True)

    def forward(self, x):
        residual = self.shortcut(x)
        x = self.conv1(x)
        x = self.conv2(x)
        x = self.se(x)
        x = self.dropout(x)
        return self.act(x + residual)


class SimpleCNN(nn.Module):
    def __init__(self, num_classes, dropout=0.35):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True),
            nn.Conv2d(32, 32, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            nn.Conv2d(32, 64, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.Conv2d(64, 64, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            nn.Conv2d(64, 128, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
            nn.Conv2d(128, 128, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            nn.Conv2d(128, 256, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(256),
            nn.ReLU(inplace=True),
            nn.Conv2d(256, 256, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(256),
            nn.ReLU(inplace=True),
            nn.AdaptiveAvgPool2d((1, 1)),
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(dropout),
            nn.Linear(256, 128),
            nn.ReLU(inplace=True),
            nn.Dropout(dropout),
            nn.Linear(128, num_classes),
        )

    def forward(self, x):
        return self.classifier(self.features(x))


class OptimizedBaselineCNN(nn.Module):
    def __init__(self, num_classes, dropout=0.42):
        super().__init__()
        self.features = nn.Sequential(
            ConvBNAct(3, 48, stride=2),
            ResidualSEBlock(48, 64, dropout=0.03),
            ResidualSEBlock(64, 96, stride=2, dropout=0.04),
            ResidualSEBlock(96, 96, dropout=0.04),
            ResidualSEBlock(96, 160, stride=2, dropout=0.05),
            ResidualSEBlock(160, 160, dropout=0.05),
            ResidualSEBlock(160, 256, stride=2, dropout=0.06),
            ResidualSEBlock(256, 256, dropout=0.06),
            ResidualSEBlock(256, 512, stride=2, dropout=0.07),
            ResidualSEBlock(512, 512, dropout=0.07),
        )
        self.avg_pool = nn.AdaptiveAvgPool2d((1, 1))
        self.max_pool = nn.AdaptiveMaxPool2d((1, 1))
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(dropout),
            nn.Linear(1024, 384),
            nn.BatchNorm1d(384),
            nn.SiLU(inplace=True),
            nn.Dropout(dropout / 2),
            nn.Linear(384, 128),
            nn.LayerNorm(128),
            nn.SiLU(inplace=True),
            nn.Dropout(dropout / 3),
            nn.Linear(128, num_classes),
        )

    def forward(self, x):
        x = self.features(x)
        x = torch.cat([self.avg_pool(x), self.max_pool(x)], dim=1)
        return self.classifier(x)


def set_parameter_requires_grad(model, requires_grad):
    for p in model.parameters():
        p.requires_grad = requires_grad

def make_transfer_head(in_features, num_classes, dropout=0.22):
    hidden = int(np.clip(in_features // 2, 512, 1536))
    hidden2 = max(hidden // 2, 256)
    return nn.Sequential(
        nn.Dropout(dropout),
        nn.Linear(in_features, hidden),
        nn.BatchNorm1d(hidden),
        nn.SiLU(inplace=True),
        nn.Dropout(dropout / 2),
        nn.Linear(hidden, hidden2),
        nn.LayerNorm(hidden2),
        nn.SiLU(inplace=True),
        nn.Dropout(dropout / 3),
        nn.Linear(hidden2, num_classes),
    )

def build_transfer_model(model_name, num_classes, dropout=0.22, freeze_backbone=True, pretrained=True):
    model_name = model_name.lower()

    if model_name == "resnet50":
        weights = models.ResNet50_Weights.DEFAULT if pretrained else None
        model = models.resnet50(weights=weights)
        in_features = model.fc.in_features
        model.fc = make_transfer_head(in_features, num_classes, dropout)
        head_modules = [model.fc]

    elif model_name == "densenet121":
        weights = models.DenseNet121_Weights.DEFAULT if pretrained else None
        model = models.densenet121(weights=weights)
        in_features = model.classifier.in_features
        model.classifier = make_transfer_head(in_features, num_classes, dropout)
        head_modules = [model.classifier]

    elif model_name == "efficientnet_b0":
        weights = models.EfficientNet_B0_Weights.DEFAULT if pretrained else None
        model = models.efficientnet_b0(weights=weights)
        in_features = model.classifier[1].in_features
        model.classifier = make_transfer_head(in_features, num_classes, dropout)
        head_modules = [model.classifier]

    elif model_name == "mobilenet_v2":
        weights = models.MobileNet_V2_Weights.DEFAULT if pretrained else None
        model = models.mobilenet_v2(weights=weights)
        in_features = model.classifier[1].in_features
        model.classifier = make_transfer_head(in_features, num_classes, dropout)
        head_modules = [model.classifier]

    else:
        raise ValueError(f"Unknown model: {model_name}")

    if freeze_backbone:
        set_parameter_requires_grad(model, False)
        for module in head_modules:
            for p in module.parameters():
                p.requires_grad = True
    return model

def unfreeze_for_finetuning(model, model_name, depth="last"):
    model_name = model_name.lower()
    set_parameter_requires_grad(model, False)

    if depth == "all":
        set_parameter_requires_grad(model, True)
        return model

    modules_by_depth = {}
    if model_name == "resnet50":
        modules_by_depth = {
            "last": [model.layer4, model.fc],
            "deeper": [model.layer3, model.layer4, model.fc],
        }
    elif model_name == "densenet121":
        modules_by_depth = {
            "last": [model.features.denseblock4, model.features.norm5, model.classifier],
            "deeper": [model.features.transition3, model.features.denseblock4, model.features.norm5, model.classifier],
        }
    elif model_name == "efficientnet_b0":
        modules_by_depth = {
            "last": [model.features[-2], model.features[-1], model.classifier],
            "deeper": [model.features[-4], model.features[-3], model.features[-2], model.features[-1], model.classifier],
        }
    elif model_name == "mobilenet_v2":
        modules_by_depth = {
            "last": [model.features[-3], model.features[-2], model.features[-1], model.classifier],
            "deeper": [model.features[-7:], model.classifier],
        }
    else:
        raise ValueError(f"Unknown model for fine-tuning: {model_name}")

    if depth not in modules_by_depth:
        raise ValueError(f"Unsupported fine-tuning depth '{depth}' for {model_name}")

    for module in modules_by_depth[depth]:
        for p in module.parameters():
            p.requires_grad = True
    return model

def make_discriminative_optimizer(model, head_lr, backbone_lr, weight_decay):
    head_keywords = ["fc", "classifier"]
    head_params = []
    backbone_params = []
    for name, p in model.named_parameters():
        if not p.requires_grad:
            continue
        if any(key in name for key in head_keywords):
            head_params.append(p)
        else:
            backbone_params.append(p)

    param_groups = []
    if backbone_params:
        param_groups.append({"params": backbone_params, "lr": backbone_lr})
    if head_params:
        param_groups.append({"params": head_params, "lr": head_lr})
    return torch.optim.AdamW(param_groups, weight_decay=weight_decay)

def build_model_from_family(family, num_classes):
    if family in {"simple_cnn", "baseline_cnn"}:
        return SimpleCNN(num_classes)
    if family == "optimized_baseline_cnn":
        return OptimizedBaselineCNN(num_classes)
    return build_transfer_model(family, num_classes, freeze_backbone=False, pretrained=False)
```

# 8. Approach 1: Simple Baseline CNN Experiments

This section evaluated the custom `SimpleCNN` architecture in two variants. The first model was trained from scratch using only deterministic preprocessing, while the second used training-time data augmentation and test-time augmentation to examine whether these techniques improved generalization.


## 8.1. Approach 1a – Baseline CNN Trained from Scratch

The `SimpleCNN` model was trained from scratch and served as the main reference point for the remaining experiments. It used deterministic preprocessing and did not include training-time augmentation, MixUp, residual blocks, Squeeze-and-Excitation attention, an additional polishing stage, or test-time augmentation.

The model was optimized with AdamW, class-weighted loss, learning-rate scheduling, early stopping, and validation-based checkpoint selection. If a saved checkpoint was available, it was loaded and evaluated instead of repeating the training process; otherwise, the model was trained, evaluated on the test set, registered, and saved together with its training history and results.

# 9. Approach 2: Optimized Baseline CNN + Augmentation + TTA

This section evaluated an optimized CNN trained entirely from scratch. The approach combined a stronger custom architecture with data augmentation, MixUp, a two-stage training procedure, and test-time augmentation.

The main training stage used X-ray-safe augmentation, weighted focal loss, label smoothing, MixUp regularization, AdamW optimization, learning-rate scheduling, early stopping, and validation-based checkpoint selection. A second low-learning-rate polishing stage used gentler augmentation and cross-entropy loss to refine the model without strongly modifying the learned features.

The best checkpoint from the main or polishing stage was selected according to validation Macro F1-score. The selected model was then evaluated using TTA probability averaging, registered in the experiment results, and saved together with its metrics and complete training history; an existing checkpoint was loaded instead when available.


```python
clear_memory()

optimized_baseline_state_key = "approach_2_optimized_baseline_cnn_aug_tta"
optimized_approach_name = "2. Optimized Baseline CNN + augmentation + TTA"
drop_experiment_result(optimized_approach_name)
model_state_registry.pop(optimized_baseline_state_key, None)

optimized_checkpoint_path = find_existing_checkpoint(
    optimized_baseline_state_key,
    extra_filenames=[CONFIG.get("optimized_baseline_checkpoint")],
)
load_optimized_checkpoint = (
    CONFIG.get("load_checkpoints_first", True)
    and CONFIG.get("load_optimized_baseline_checkpoint", True)
    and optimized_checkpoint_path is not None
)

if load_optimized_checkpoint:
    optimized_baseline_model = OptimizedBaselineCNN(NUM_CLASSES).to(device)
    optimized_baseline_model, optimized_baseline_history, optimized_baseline_training_state, optimized_baseline_metrics = load_checkpoint_experiment(
        checkpoint_path=optimized_checkpoint_path,
        model=optimized_baseline_model,
        state_key=optimized_baseline_state_key,
        model_family="optimized_baseline_cnn",
        approach_name=optimized_approach_name,
        history_title="Optimized Baseline CNN + augmentation + TTA",
        eval_title="Approach 2 - Optimized Baseline CNN + augmentation + TTA",
        eval_loader=test_loader,
        eval_tta=CONFIG.get("baseline_use_tta", True),
        notes="Optimized scratch Baseline CNN with X-ray-safe augmentation, MixUp, polish stage and TTA evaluation.",
        stage_rename={
            "baseline main": "optimized baseline main",
            "baseline polish": "optimized baseline polish",
        },
        stage_total_epochs={
            "optimized baseline main": CONFIG["epochs_baseline"],
            "optimized baseline polish": CONFIG["epochs_baseline_polish"],
        },
        checkpoint_tags={
            "optimized baseline main": optimized_baseline_state_key,
            "optimized baseline polish": f"{optimized_baseline_state_key}_polish",
        },
    )
else:
    if CONFIG.get("load_checkpoints_first", True):
        print("Optimized baseline checkpoint not found; training Approach 2 from scratch.")
    train_loader_optimized, val_loader, test_loader, class_names, class_weights = make_loaders(
        baseline_train_transform,
        max_samples_per_class=CONFIG["max_samples_per_class"],
        use_weighted_sampler=CONFIG["use_weighted_sampler"],
    )

    optimized_baseline_model = OptimizedBaselineCNN(NUM_CLASSES).to(device)
    display_model_profile(optimized_baseline_model)

    criterion = make_criterion(
        class_weights,
        label_smoothing=CONFIG["baseline_label_smoothing"],
        loss_name="focal",
        focal_gamma=CONFIG["baseline_focal_gamma"],
    )
    optimizer = torch.optim.AdamW(
        optimized_baseline_model.parameters(),
        lr=CONFIG["lr_baseline"],
        weight_decay=CONFIG["weight_decay"],
    )

    optimized_baseline_model, optimized_history_main, optimized_state_main = fit_model(
        optimized_baseline_model,
        train_loader_optimized,
        val_loader,
        criterion,
        optimizer,
        epochs=CONFIG["epochs_baseline"],
        phase_name="optimized baseline main",
        checkpoint_tag=optimized_baseline_state_key,
        early_stopping_patience=CONFIG["baseline_early_stopping_patience"],
        min_epochs_before_stopping=CONFIG["baseline_min_epochs_before_stopping"],
        scheduler_patience=CONFIG["baseline_scheduler_patience"],
        mixup_alpha=CONFIG["baseline_mixup_alpha"],
        mixup_prob=CONFIG["baseline_mixup_prob"],
    )

    train_loader_optimized_polish, val_loader, test_loader, class_names, class_weights = make_loaders(
        baseline_polish_transform,
        max_samples_per_class=CONFIG["max_samples_per_class"],
        use_weighted_sampler=CONFIG["use_weighted_sampler"],
    )
    optimized_baseline_model.load_state_dict(optimized_state_main["model_state"])
    polish_criterion = make_criterion(
        class_weights,
        label_smoothing=CONFIG["baseline_polish_label_smoothing"],
        loss_name="cross_entropy",
    )
    polish_optimizer = torch.optim.AdamW(
        optimized_baseline_model.parameters(),
        lr=CONFIG["lr_baseline_polish"],
        weight_decay=CONFIG["weight_decay"],
    )
    optimized_baseline_model, optimized_history_polish, optimized_state_polish = fit_model(
        optimized_baseline_model,
        train_loader_optimized_polish,
        val_loader,
        polish_criterion,
        polish_optimizer,
        epochs=CONFIG["epochs_baseline_polish"],
        phase_name="optimized baseline polish",
        checkpoint_tag=f"{optimized_baseline_state_key}_polish",
        early_stopping_patience=max(8, CONFIG["baseline_early_stopping_patience"] // 2),
        min_epochs_before_stopping=min(10, CONFIG["epochs_baseline_polish"]),
        scheduler_patience=2,
    )

    optimized_baseline_training_state = choose_better_state(
        optimized_baseline_model,
        optimized_state_main,
        optimized_state_polish,
    )
    optimized_baseline_history = merge_histories(optimized_history_main, optimized_history_polish)
    register_model_state(
        optimized_baseline_state_key,
        optimized_baseline_training_state,
        "optimized_baseline_cnn",
    )
    plot_history(optimized_baseline_history, "Optimized Baseline CNN + augmentation + TTA")

    if CONFIG.get("baseline_use_tta", True):
        optimized_baseline_metrics, _, _ = evaluate_model_tta(
            optimized_baseline_model,
            "test",
            class_names,
            "Approach 2 - Optimized Baseline CNN + augmentation + TTA",
        )
    else:
        optimized_baseline_metrics, _, _ = evaluate_model(
            optimized_baseline_model,
            test_loader,
            class_names,
            "Approach 2 - Optimized Baseline CNN + augmentation",
        )

    add_experiment_result(
        optimized_approach_name,
        optimized_baseline_model,
        optimized_baseline_metrics,
        optimized_baseline_history,
        optimized_baseline_state_key,
        model_family="optimized_baseline_cnn",
        notes="Optimized scratch Baseline CNN with X-ray-safe augmentation, MixUp, polish stage and TTA evaluation.",
        best_val_score=optimized_baseline_training_state["best_score"],
    )
    save_model_checkpoint(
        optimized_baseline_state_key,
        optimized_baseline_model,
        optimized_baseline_training_state,
        metrics=optimized_baseline_metrics,
        history_df=optimized_baseline_history,
    )

baseline_aug_model = optimized_baseline_model
baseline_aug_history = optimized_baseline_history
baseline_aug_training_state = optimized_baseline_training_state
baseline_aug_metrics = optimized_baseline_metrics
baseline_aug_state_key = optimized_baseline_state_key

save_results_csv()
display(results_table())
```

# 10. Approach 3: Transfer Learning

This section compared four ImageNet-pretrained CNN architectures adapted to the six-class chest X-ray classification task. Each model was trained with the same validation-based protocol so that their predictive quality and efficiency could be compared fairly.

The transfer-learning pipeline was prepared for ResNet50, DenseNet121, EfficientNet-B0, and MobileNetV2. Separate dictionaries stored the checkpoint key, complete training history, and best validation Macro F1-score obtained by each architecture.

Each experiment followed three stages:

1. **Classification-head warm-up** – the pretrained backbone was frozen and only the new classification head was trained.
2. **Last-block adaptation** – the final backbone block was unfrozen and optimized with a smaller learning rate than the classification head.
3. **Low-learning-rate polishing** – the adapted model was refined using gentler augmentation and reduced learning rates.

The models used ImageNet weights, a shared multi-layer classification head, mild X-ray-safe augmentation, weighted focal loss, label smoothing, AdamW optimization, learning-rate scheduling, early stopping, and discriminative learning rates. After every stage, the checkpoint with the highest validation Macro F1-score was retained, and the final model was evaluated on the test set, registered, and saved.

If an existing checkpoint was available, it was loaded and evaluated instead of repeating training. This also allowed the stored epoch history and model-selection process to be reproduced.


```python
transfer_model_names = list(CONFIG["transfer_models"])

transfer_state_keys = {}
transfer_histories = {}
transfer_val_scores = {}

train_loader_transfer, val_loader, test_loader, class_names, class_weights = make_loaders(
    transfer_train_transform,
    max_samples_per_class=CONFIG["max_samples_per_class"],
    use_weighted_sampler=CONFIG["transfer_use_weighted_sampler"],
)
train_loader_transfer_polish, _, _, _, _ = make_loaders(
    transfer_polish_transform,
    max_samples_per_class=CONFIG["max_samples_per_class"],
    use_weighted_sampler=CONFIG["transfer_use_weighted_sampler"],
)

def remove_previous_experiment_result(approach_name):
    drop_experiment_result(approach_name)

def run_transfer_experiment(model_name):
    model_name = model_name.lower()
    if model_name not in transfer_model_names:
        raise ValueError(f"{model_name} is not enabled in CONFIG['transfer_models'].")

    print("\n" + "=" * 80)
    print(f"Transfer learning: {model_name}")
    print("=" * 80)
    clear_memory()

    state_key = f"approach_3_transfer_{model_name}"
    approach_name = f"3. Transfer learning - {model_name}"
    checkpoint_path = find_existing_checkpoint(state_key)

    if CONFIG.get("load_checkpoints_first", True) and checkpoint_path is not None:
        model = build_transfer_model(model_name, NUM_CLASSES, freeze_backbone=True, pretrained=False).to(device)
        model = unfreeze_for_finetuning(model, model_name, depth="last").to(device)
        model, history, training_state, metrics = load_checkpoint_experiment(
            checkpoint_path=checkpoint_path,
            model=model,
            state_key=state_key,
            model_family=model_name,
            approach_name=approach_name,
            history_title=f"Transfer learning - {model_name}",
            eval_title=f"Approach 3 - Transfer learning: {model_name}",
            eval_loader=test_loader,
            notes="ImageNet backbone with stronger classifier head, mild X-ray augmentation, focal loss, last-block adaptation, and low-LR polish.",
            stage_total_epochs={
                "head warm-up": CONFIG["epochs_transfer_head"],
                "last block adaptation": CONFIG["epochs_transfer_unfreeze"],
                "last block low-lr polish": CONFIG.get("epochs_transfer_polish", 0),
            },
            checkpoint_tags={
                "head warm-up": f"approach_3_{model_name}_head_warmup",
                "last block adaptation": f"approach_3_{model_name}_last_block_adaptation",
                "last block low-lr polish": f"approach_3_{model_name}_last_block_polish",
            },
        )
    else:
        if CONFIG.get("load_checkpoints_first", True):
            print(f"No checkpoint found for {state_key}; training transfer model from scratch/fine-tuning ImageNet weights.")
        model = build_transfer_model(model_name, NUM_CLASSES, freeze_backbone=True).to(device)
        display_model_profile(model)
        criterion = make_criterion(
            class_weights,
            label_smoothing=CONFIG.get("transfer_label_smoothing", CONFIG["label_smoothing"]),
            loss_name="focal",
            focal_gamma=CONFIG.get("transfer_focal_gamma", 1.0),
        )

        transfer_fit_kwargs = {
            "early_stopping_patience": CONFIG.get("transfer_early_stopping_patience", CONFIG["early_stopping_patience"]),
            "min_epochs_before_stopping": CONFIG.get("transfer_min_epochs_before_stopping", CONFIG["min_epochs_before_stopping"]),
            "scheduler_patience": CONFIG.get("transfer_scheduler_patience", 2),
        }

        head_optimizer = torch.optim.AdamW(
            [p for p in model.parameters() if p.requires_grad],
            lr=CONFIG.get("lr_transfer_head_warmup", CONFIG["lr_transfer_head"]),
            weight_decay=CONFIG["weight_decay"],
        )
        model, head_history, head_state = fit_model(
            model,
            train_loader_transfer,
            val_loader,
            criterion,
            head_optimizer,
            epochs=CONFIG["epochs_transfer_head"],
            phase_name="head warm-up",
            checkpoint_tag=f"approach_3_{model_name}_head_warmup",
            **transfer_fit_kwargs,
        )

        model = unfreeze_for_finetuning(model, model_name, depth="last").to(device)
        adapt_optimizer = make_discriminative_optimizer(
            model,
            head_lr=CONFIG.get("lr_transfer_adapt_head", CONFIG["lr_transfer_head"]),
            backbone_lr=CONFIG.get("lr_transfer_adapt_backbone", CONFIG["lr_transfer_backbone"]),
            weight_decay=CONFIG["weight_decay"],
        )
        model, adapt_history, adapt_state = fit_model(
            model,
            train_loader_transfer,
            val_loader,
            criterion,
            adapt_optimizer,
            epochs=CONFIG["epochs_transfer_unfreeze"],
            phase_name="last block adaptation",
            checkpoint_tag=f"approach_3_{model_name}_last_block_adaptation",
            **transfer_fit_kwargs,
        )

        best_after_adapt = choose_better_state(model, head_state, adapt_state)

        model = unfreeze_for_finetuning(model, model_name, depth="last").to(device)
        polish_optimizer = make_discriminative_optimizer(
            model,
            head_lr=CONFIG.get("lr_transfer_polish_head", CONFIG.get("lr_transfer_adapt_head", CONFIG["lr_transfer_head"])),
            backbone_lr=CONFIG.get("lr_transfer_polish_backbone", CONFIG.get("lr_transfer_adapt_backbone", CONFIG["lr_transfer_backbone"])),
            weight_decay=CONFIG["weight_decay"],
        )
        model, polish_history, polish_state = fit_model(
            model,
            train_loader_transfer_polish,
            val_loader,
            criterion,
            polish_optimizer,
            epochs=CONFIG.get("epochs_transfer_polish", 0),
            phase_name="last block low-lr polish",
            checkpoint_tag=f"approach_3_{model_name}_last_block_polish",
            **transfer_fit_kwargs,
        )

        training_state = choose_better_state(model, best_after_adapt, polish_state)
        history = merge_histories(head_history, adapt_history, polish_history)
        register_model_state(state_key, training_state, model_name)
        plot_history(history, f"Transfer learning - {model_name}")
        metrics, _, _ = evaluate_model(model, test_loader, class_names, f"Approach 3 - Transfer learning: {model_name}")

        remove_previous_experiment_result(approach_name)
        add_experiment_result(
            approach_name,
            model,
            metrics,
            history,
            state_key,
            model_family=model_name,
            notes="ImageNet backbone with stronger classifier head, mild X-ray augmentation, focal loss, last-block adaptation, and low-LR polish.",
            best_val_score=training_state["best_score"],
        )
        save_model_checkpoint(
            state_key,
            model,
            training_state,
            metrics=metrics,
            history_df=history,
        )

    transfer_state_keys[model_name] = state_key
    transfer_histories[model_name] = history
    transfer_val_scores[model_name] = float(training_state["best_score"])
    save_results_csv()

    del model
    clear_memory()
    display(results_table()[results_table()["approach"].str.startswith("3.")])
    return state_key

print("Transfer-learning setup ready. Run one model cell below at a time.")
print("Enabled transfer models:", transfer_model_names)
```

The transfer-learning function was executed separately for:

* ResNet50,
* DenseNet121,
* EfficientNet-B0,
* MobileNetV2.

Running each architecture in an individual cell reduced GPU memory usage and allowed the models to be trained, loaded, evaluated, and saved independently. After each run, its validation score, test metrics, model state, training history, and efficiency measurements were added to the common experiment registry.

# 11. Approach 4: Fine-Tuning the Best Transfer Model

This section further adapted the best transfer-learning model selected in Approach 3. Fine-tuning was restricted to the validation winner to avoid unnecessary experiments with weaker architectures.


The model with the highest transfer-learning validation Macro F1-score was selected automatically, and only this architecture was allowed to proceed to fine-tuning. Its best saved transfer checkpoint was used as the starting point rather than initializing the model again.

For ResNet50, a conservative refinement strategy was applied:

* the final residual block and classification head were unfrozen,
* the pretrained backbone received a much smaller learning rate than the classifier,
* mild fine-tuning augmentation was used,
* label smoothing, AdamW, learning-rate scheduling, and early stopping were applied,
* the new checkpoint was compared with the original transfer-learning state using validation Macro F1-score.

An optional deeper refinement of `layer3` was supported, but it was performed only when its configured number of epochs was greater than zero. If fine-tuning reduced validation performance, the original transfer checkpoint was preserved.

The selected state was evaluated on the test set, registered as Approach 4, and saved together with its metrics and training history. An existing checkpoint was loaded instead when available.


```python
if not transfer_val_scores:
    raise RuntimeError("Run at least one transfer-learning model before fine-tuning.")

if "best_transfer_name" not in globals() or best_transfer_name not in transfer_state_keys:
    best_transfer_name = max(transfer_val_scores, key=transfer_val_scores.get)
    best_transfer_state_key = transfer_state_keys[best_transfer_name]

best_transfer_name = best_transfer_name.lower()
best_transfer_state_key = transfer_state_keys[best_transfer_name]
ft_candidate_names = [best_transfer_name]

print("Fine-tuning will run only for the best transfer model:")
print(f"- {best_transfer_name} | validation macro F1={transfer_val_scores[best_transfer_name]:.4f}")

ft_state_keys = {}
ft_histories = {}
ft_val_scores = {}

train_loader_finetune, val_loader, test_loader, class_names, class_weights = make_loaders(
    finetune_transform,
    max_samples_per_class=CONFIG["max_samples_per_class"],
    use_weighted_sampler=CONFIG["transfer_use_weighted_sampler"],
)

def remove_previous_finetune_results():
    experiment_results[:] = [
        row for row in experiment_results
        if not str(row.get("approach", "")).startswith("4.")
    ]

def run_finetune_experiment(model_name=None):
    model_name = (model_name or best_transfer_name).lower()

    if model_name != best_transfer_name:
        raise RuntimeError(
            f"Fine-tuning is restricted to the selected best transfer model ({best_transfer_name}). "
            "Rerun the transfer selection cell if you want to change the winner."
        )
    if model_name not in transfer_state_keys:
        raise RuntimeError(f"Run the transfer-learning cell for {model_name} before fine-tuning it.")

    print("\n" + "=" * 80)
    print(f"Fine-tuning selected transfer winner: {model_name}")
    print("=" * 80)
    clear_memory()

    ft_state_key = f"approach_4_finetuned_{model_name}"
    approach_name = f"4. Fine-tuned best transfer model - {model_name}"
    is_resnet = model_name == "resnet50"
    checkpoint_path = find_existing_checkpoint(ft_state_key)

    if CONFIG.get("load_checkpoints_first", True) and checkpoint_path is not None:
        ft_model = build_transfer_model(model_name, NUM_CLASSES, freeze_backbone=True, pretrained=False).to(device)
        ft_model = unfreeze_for_finetuning(ft_model, model_name, depth="deeper").to(device)
        ft_note = (
            "Validation-guarded ResNet50 fine-tuning with conservative last-block and layer3 polish."
            if is_resnet
            else "Validation-guarded deeper fine-tuning of the selected transfer winner."
        )
        stage_total_epochs = (
            {
                "resnet last-block refinement": CONFIG["resnet_epochs_last_block"],
                "resnet layer3 low-lr polish": CONFIG["resnet_epochs_layer3_polish"],
            }
            if is_resnet
            else {"selected-model deeper fine-tuning": CONFIG["epochs_finetune"]}
        )
        checkpoint_tags = (
            {
                "resnet last-block refinement": f"approach_4_{model_name}_last_block_refinement",
                "resnet layer3 low-lr polish": f"approach_4_{model_name}_layer3_polish",
            }
            if is_resnet
            else {"selected-model deeper fine-tuning": f"approach_4_{model_name}_deeper_finetuning"}
        )
        remove_previous_finetune_results()
        ft_model, ft_history, ft_training_state, ft_metrics = load_checkpoint_experiment(
            checkpoint_path=checkpoint_path,
            model=ft_model,
            state_key=ft_state_key,
            model_family=model_name,
            approach_name=approach_name,
            history_title=f"Fine-tuned selected model - {model_name}",
            eval_title=f"Approach 4 - Fine-tuned selected transfer model: {model_name}",
            eval_loader=test_loader,
            notes=ft_note,
            stage_total_epochs=stage_total_epochs,
            checkpoint_tags=checkpoint_tags,
        )
    else:
        if CONFIG.get("load_checkpoints_first", True):
            print(f"No checkpoint found for {ft_state_key}; running fine-tuning.")
        ft_model = build_transfer_model(model_name, NUM_CLASSES, freeze_backbone=True, pretrained=False).to(device)
        ft_model.load_state_dict(model_state_registry[transfer_state_keys[model_name]]["model_state"])
        starting_state = training_state_from_registry(transfer_state_keys[model_name])
        display_model_profile(ft_model)

        if is_resnet:
            criterion = make_criterion(
                class_weights,
                label_smoothing=CONFIG["resnet_label_smoothing"],
            )

            ft_model = unfreeze_for_finetuning(ft_model, model_name, depth="last").to(device)
            last_optimizer = make_discriminative_optimizer(
                ft_model,
                head_lr=CONFIG["resnet_lr_last_head"],
                backbone_lr=CONFIG["resnet_lr_last_backbone"],
                weight_decay=CONFIG["weight_decay"],
            )
            ft_model, last_history, last_state = fit_model(
                ft_model,
                train_loader_finetune,
                val_loader,
                criterion,
                last_optimizer,
                epochs=CONFIG["resnet_epochs_last_block"],
                phase_name="resnet last-block refinement",
                checkpoint_tag=f"approach_4_{model_name}_last_block_refinement",
                early_stopping_patience=8,
                min_epochs_before_stopping=12,
                scheduler_patience=3,
            )
            best_after_last = choose_better_state(ft_model, starting_state, last_state)

            if CONFIG["resnet_epochs_layer3_polish"] > 0:
              ft_model = unfreeze_for_finetuning(ft_model, model_name, depth="deeper").to(device)
              polish_optimizer = make_discriminative_optimizer(
                  ft_model,
                  head_lr=CONFIG["resnet_lr_layer3_head"],
                  backbone_lr=CONFIG["resnet_lr_layer3_backbone"],
                  weight_decay=CONFIG["weight_decay"],
              )
              ft_model, polish_history, polish_state = fit_model(
                  ft_model,
                  train_loader_finetune,
                  val_loader,
                  criterion,
                  polish_optimizer,
                  epochs=CONFIG["resnet_epochs_layer3_polish"],
                  phase_name="resnet layer3 low-lr polish",
                  checkpoint_tag=f"approach_4_{model_name}_layer3_polish",
              )
              ft_training_state = choose_better_state(ft_model, best_after_last, polish_state)
              ft_history = merge_histories(last_history, polish_history)
              ft_note = "Validation-guarded ResNet50 fine-tuning with conservative last-block and optional layer3 polish."
            else:
              ft_model.load_state_dict(best_after_last["model_state"])
              ft_training_state = best_after_last
              ft_history = last_history
              ft_note = "Validation-guarded ResNet50 micro fine-tuning with conservative last-block refinement only."

        else:
            ft_model = unfreeze_for_finetuning(ft_model, model_name, depth="deeper").to(device)
            criterion = make_criterion(class_weights)
            optimizer = make_discriminative_optimizer(
                ft_model,
                head_lr=CONFIG["lr_finetune_head"],
                backbone_lr=CONFIG["lr_finetune_backbone"],
                weight_decay=CONFIG["weight_decay"],
            )
            ft_model, ft_history, candidate_state = fit_model(
                ft_model,
                train_loader_finetune,
                val_loader,
                criterion,
                optimizer,
                epochs=CONFIG["epochs_finetune"],
                phase_name="selected-model deeper fine-tuning",
                checkpoint_tag=f"approach_4_{model_name}_deeper_finetuning",
            )
            ft_training_state = choose_better_state(ft_model, starting_state, candidate_state)
            ft_note = "Validation-guarded deeper fine-tuning of the selected transfer winner."

        if ft_training_state["best_score"] < starting_state["best_score"]:
            print("Fine-tuning did not beat the transfer checkpoint; keeping the transfer state for this approach.")

        remove_previous_finetune_results()
        register_model_state(ft_state_key, ft_training_state, model_name)
        plot_history(ft_history, f"Fine-tuned selected model - {model_name}")

        ft_metrics, _, _ = evaluate_model(
            ft_model,
            test_loader,
            class_names,
            f"Approach 4 - Fine-tuned selected transfer model: {model_name}",
        )

        add_experiment_result(
            approach_name,
            ft_model,
            ft_metrics,
            ft_history,
            ft_state_key,
            model_family=model_name,
            notes=ft_note,
            best_val_score=ft_training_state["best_score"],
        )

        save_model_checkpoint(
            ft_state_key,
            ft_model,
            ft_training_state,
            metrics=ft_metrics,
            history_df=ft_history,
        )

    save_results_csv()

    ft_state_keys[model_name] = ft_state_key
    ft_histories[model_name] = ft_history
    ft_val_scores[model_name] = float(ft_training_state["best_score"])

    del ft_model
    clear_memory()

    display(results_table()[results_table()["approach"].str.startswith("4.")])
    return ft_state_key

print("Fine-tuning setup ready. Run the single selected-model fine-tuning cell below.")
```

This cell executed the fine-tuning pipeline for the selected best transfer model and stored the resulting checkpoint key.

```python
best_transfer_ft_state_key = run_finetune_experiment()
```

# 13. Approach 6: Quality-First Pruning for Efficiency

This section investigated whether the strongest eligible ResNet50 model could be made more sparse while preserving its classification quality. A quality-first pruning strategy was used, so validation Macro F1-score remained the primary selection criterion.


The strongest model from Approaches 4 and 5 was selected automatically using validation Macro F1-score. Global unstructured L1 pruning was then applied to the weights of all convolutional and linear layers.

Several pruning ratios were supported. For every candidate:

* the weights with the smallest absolute values were removed,
* pruning masks remained active during low-learning-rate recovery training,
* only the final backbone block and classification head were adapted,
* validation Macro F1-score and final sparsity were measured,
* early stopping and checkpointing were applied.

The candidate with the highest validation Macro F1-score was selected. Sparsity was used only as a secondary criterion for candidates with nearly equal validation performance.


```python
def collect_prunable_modules(model):
    return [
        (module, "weight")
        for module in model.modules()
        if isinstance(module, (nn.Conv2d, nn.Linear))
    ]

def apply_global_l1_pruning(model, amount):
    modules_to_prune = collect_prunable_modules(model)
    prune.global_unstructured(
        modules_to_prune,
        pruning_method=prune.L1Unstructured,
        amount=amount,
    )
    return modules_to_prune

def remove_pruning_reparam(modules_to_prune):
    for module, name in modules_to_prune:
        prune.remove(module, name)

def pruning_amount_label(amount):
    label = f"{amount * 100:.1f}".rstrip("0").rstrip(".")
    return label.replace(".", "p") + "pct"

current_results = results_table()
fine_tuned_rows = current_results[current_results["approach"].str.startswith(("4.", "5."))].copy()
if fine_tuned_rows.empty:
    raise RuntimeError("Run fine-tuning or extra-soft augmentation before pruning.")

best_single_row = sort_for_validation_selection(fine_tuned_rows, top_k=1).iloc[0]
source_state_key = best_single_row["state_key"]
source_family = best_single_row["model_family"]
source_val_score = float(best_single_row["best_val_macro_f1"])
print("Pruning source model selected by validation macro F1:")
display(best_single_row.to_frame().T)

experiment_results[:] = [
    row for row in experiment_results
    if not str(row.get("approach", "")).startswith("6.")
]

existing_pruned_paths = sorted(
    SAVE_DIR.glob(f"{safe_checkpoint_name(f'approach_6_pruned_{source_family}_')}*.pt"),
    key=lambda p: p.stat().st_mtime,
    reverse=True,
)

loaded_pruned_checkpoint = False
if CONFIG.get("load_checkpoints_first", True) and existing_pruned_paths:
    checkpoint_path = existing_pruned_paths[0]
    checkpoint = load_saved_checkpoint(checkpoint_path)
    prune_state_key = checkpoint.get("state_key") or checkpoint_path.stem
    prune_amount = None
    for amount in CONFIG["pruning_candidates"]:
        if pruning_amount_label(amount) in prune_state_key:
            prune_amount = amount
            break
    amount_label = pruning_amount_label(prune_amount) if prune_amount is not None else "checkpoint"
    approach_name = (
        f"6. Quality-first pruned best model - {source_family} ({prune_amount:.1%})"
        if prune_amount is not None
        else f"6. Quality-first pruned best model - {source_family}"
    )
    drop_experiment_result(approach_name)
    model_state_registry.pop(prune_state_key, None)
    pruned_model = build_model_from_family(source_family, NUM_CLASSES).to(device)
    pruned_model = unfreeze_for_finetuning(pruned_model, source_family, depth="last").to(device)

    state_dict = checkpoint_model_state_dict(checkpoint)
    if state_dict is None:
        raise RuntimeError(f"No model weights found in checkpoint: {checkpoint_path}")
    pruned_model.load_state_dict(state_dict)
    pruned_model.eval()
    display_model_profile(pruned_model)

    selected_history = checkpoint_history_df(checkpoint)
    replay_checkpoint_history(
        selected_history,
        stage_total_epochs={
            f"quality-first prune recovery {prune_amount:.1%}": CONFIG["epochs_prune_recovery"]
        } if prune_amount is not None else None,
        checkpoint_tags={
            f"quality-first prune recovery {prune_amount:.1%}": f"approach_6_{source_family}_prune_{amount_label}_recovery"
        } if prune_amount is not None else None,
        default_checkpoint_tag=f"approach_6_{source_family}_prune_{amount_label}_recovery",
    )
    plot_history(
        selected_history,
        f"Pruned {source_family} ({prune_amount:.1%})" if prune_amount is not None else f"Pruned {source_family}",
    )
    pruned_metrics, _, _ = evaluate_model(
        pruned_model,
        test_loader,
        class_names,
        f"Approach 6 - Quality-first pruned {source_family} ({prune_amount:.1%})" if prune_amount is not None else f"Approach 6 - Quality-first pruned {source_family}",
    )
    prune_training_state = checkpoint_training_state(checkpoint, pruned_model, selected_history)
    register_model_state(prune_state_key, prune_training_state, source_family)
    best_score = prune_training_state["best_score"]
    final_sparsity = global_sparsity(pruned_model)
    add_experiment_result(
        approach_name,
        pruned_model,
        pruned_metrics,
        selected_history,
        prune_state_key,
        model_family=source_family,
        notes=(
            f"Loaded quality-first global L1 pruned checkpoint. "
            f"Amount={prune_amount:.1%}; final sparsity={final_sparsity:.1%}."
            if prune_amount is not None
            else f"Loaded quality-first global L1 pruned checkpoint; final sparsity={final_sparsity:.1%}."
        ),
        best_val_score=best_score,
    )
    loaded_pruned_checkpoint = True

if not loaded_pruned_checkpoint:
    prune_candidates = []
    for amount in CONFIG["pruning_candidates"]:
        amount_label = pruning_amount_label(amount)
        print("\n" + "=" * 80)
        print(f"Pruning candidate: {source_family}, amount={amount:.1%}")
        print("=" * 80)
        clear_memory()

        candidate_model = build_model_from_family(source_family, NUM_CLASSES).to(device)
        candidate_model.load_state_dict(model_state_registry[source_state_key]["model_state"])
        before_sparsity = global_sparsity(candidate_model)
        modules_to_prune = apply_global_l1_pruning(candidate_model, amount=amount)
        masked_sparsity = global_sparsity(candidate_model)

        candidate_model = unfreeze_for_finetuning(
            candidate_model,
            source_family,
            depth="last"
        ).to(device)
        display_model_profile(candidate_model)

        criterion = make_criterion(
            class_weights,
            label_smoothing=CONFIG.get("pruning_label_smoothing", CONFIG["label_smoothing"])
        )

        optimizer = make_discriminative_optimizer(
            candidate_model,
            head_lr=CONFIG["lr_prune_recovery_head"],
            backbone_lr=CONFIG["lr_prune_recovery_backbone"],
            weight_decay=CONFIG["weight_decay"],
        )

        candidate_model, candidate_history, candidate_state = fit_model(
            candidate_model,
            train_loader_soft,
            val_loader,
            criterion,
            optimizer,
            epochs=CONFIG["epochs_prune_recovery"],
            phase_name=f"quality-first prune recovery {amount:.1%}",
            checkpoint_tag=f"approach_6_{source_family}_prune_{amount_label}_recovery",
        )
        candidate_model.load_state_dict(candidate_state["model_state"])
        remove_pruning_reparam(modules_to_prune)
        final_sparsity = global_sparsity(candidate_model)
        candidate_state["model_state"] = copy_state_dict_to_cpu(candidate_model)

        prune_candidates.append({
            "amount": amount,
            "amount_label": amount_label,
            "model_state": candidate_state,
            "history": candidate_history,
            "best_score": float(candidate_state["best_score"]),
            "delta_vs_source": float(candidate_state["best_score"]) - source_val_score,
            "before_sparsity": before_sparsity,
            "masked_sparsity": masked_sparsity,
            "final_sparsity": final_sparsity,
        })
        print(f"Sparsity before pruning: {before_sparsity:.2%}")
        print(f"Sparsity after pruning/recovery: {final_sparsity:.2%}")
        print(f"Best validation macro F1 after recovery: {candidate_state['best_score']:.4f}")

        del candidate_model
        clear_memory()

    candidate_report = pd.DataFrame([
        {
            "amount": c["amount"],
            "best_val_macro_f1": c["best_score"],
            "delta_vs_source_val": c["delta_vs_source"],
            "final_sparsity": c["final_sparsity"],
        }
        for c in prune_candidates
    ]).sort_values(["best_val_macro_f1", "final_sparsity"], ascending=[False, False])

    print("Pruning candidate validation summary:")
    display(candidate_report.style.format({
        "amount": "{:.1%}",
        "best_val_macro_f1": "{:.4f}",
        "delta_vs_source_val": "{:+.4f}",
        "final_sparsity": "{:.2%}",
    }))

    min_amount = CONFIG.get("pruning_min_amount", min(CONFIG["pruning_candidates"]))
    eligible_candidates = [c for c in prune_candidates if c["amount"] >= min_amount]
    if not eligible_candidates:
        raise RuntimeError("No pruning candidate satisfies CONFIG['pruning_min_amount'].")

    # Quality-first selection: maximize validation macro F1; use sparsity only among near-equal candidates.
    best_candidate_score = max(c["best_score"] for c in eligible_candidates)
    quality_margin = CONFIG.get("pruning_quality_margin", 0.0)
    near_best_candidates = [
        c for c in eligible_candidates
        if c["best_score"] >= best_candidate_score - quality_margin
    ]
    selected_prune = sorted(
        near_best_candidates,
        key=lambda c: (c["final_sparsity"], c["amount"], c["best_score"]),
        reverse=True,
    )[0]

    allowed_floor = source_val_score - CONFIG.get("pruning_drop_tolerance", 0.0)
    if selected_prune["best_score"] < allowed_floor:
        print(
            "Warning: all pruned candidates are below the configured validation tolerance. "
            "The highest-validation pruned candidate is still selected because this section must report a pruned model."
        )

    print(
        f"Selected pruning candidate: amount={selected_prune['amount']:.1%}, "
        f"val_macro_f1={selected_prune['best_score']:.4f}, "
        f"delta_vs_source={selected_prune['delta_vs_source']:+.4f}, "
        f"final_sparsity={selected_prune['final_sparsity']:.2%}"
    )

    prune_state_key = f"approach_6_pruned_{source_family}_{selected_prune['amount_label']}"
    prune_training_state = selected_prune["model_state"]
    register_model_state(prune_state_key, prune_training_state, source_family)

    pruned_model = build_model_from_family(source_family, NUM_CLASSES).to(device)
    pruned_model.load_state_dict(prune_training_state["model_state"])
    plot_history(selected_prune["history"], f"Pruned {source_family} ({selected_prune['amount']:.1%})")
    pruned_metrics, _, _ = evaluate_model(pruned_model, test_loader, class_names, f"Approach 6 - Quality-first pruned {source_family} ({selected_prune['amount']:.1%})")
    add_experiment_result(
        f"6. Quality-first pruned best model - {source_family} ({selected_prune['amount']:.1%})",
        pruned_model,
        pruned_metrics,
        selected_prune["history"],
        prune_state_key,
        model_family=source_family,
        notes=(
            f"Quality-first global L1 pruning selected by validation macro F1. "
            f"Amount={selected_prune['amount']:.1%}; final sparsity={selected_prune['final_sparsity']:.1%}; "
            f"validation delta vs source={selected_prune['delta_vs_source']:+.4f}."
        ),
        best_val_score=selected_prune["best_score"],
    )
    save_model_checkpoint(
        prune_state_key,
        pruned_model,
        prune_training_state,
        metrics=pruned_metrics,
        history_df=selected_prune["history"],
    )

save_results_csv()
display(results_table())
```

# 14. Approach 7: Validation-Greedy Model Ensemble with Test-Time Augmentation

This section combined complementary models using a validation-greedy ensemble and test-time augmentation. The objective was to improve classification robustness by averaging probability predictions from models that reduced validation errors when combined.

The candidate models were ranked using validation Macro F1-score, without using test-set results for ensemble selection. Each candidate was added sequentially, and it was retained only when it improved the ensemble validation Macro F1-score.

The implemented procedure:

* generated predictions for every candidate on the validation set,
* verified identical sample ordering across models,
* averaged class probabilities instead of predicted labels,
* selected only models that improved validation Macro F1-score,
* applied three deterministic test-time transformations,
* averaged predictions across all selected models and TTA variants,
* calculated final test metrics and confusion matrices.

This approach allowed models with different architectures and error patterns to contribute complementary information.


```python
def tta_transforms():
    return [
        eval_transform,
        transforms.Compose([
            transforms.Resize((238, 238)),
            transforms.Grayscale(num_output_channels=3),
            transforms.CenterCrop(IMG_SIZE),
            transforms.ToTensor(),
            transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
        ]),
        transforms.Compose([
            transforms.Resize((232, 232)),
            transforms.Grayscale(num_output_channels=3),
            transforms.CenterCrop(IMG_SIZE),
            transforms.ToTensor(),
            transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
        ]),
    ]

@torch.no_grad()
def predict_state_key(state_key, loader):
    model_family = model_state_registry[state_key]["model_family"]
    model = build_model_from_family(model_family, NUM_CLASSES).to(device)
    model.load_state_dict(model_state_registry[state_key]["model_state"])
    model.eval()
    y_true, y_prob = collect_predictions(model, loader)
    del model
    clear_memory()
    return y_true, y_prob

def greedy_select_ensemble(candidate_df, validation_loader):
    selected_rows = []
    selected_probs = []
    y_true_ref = None
    best_score = -np.inf

    for _, row in candidate_df.iterrows():
        y_true, y_prob = predict_state_key(row["state_key"], validation_loader)
        if y_true_ref is None:
            y_true_ref = y_true
        else:
            assert np.array_equal(y_true_ref, y_true), "Validation loader order mismatch."

        trial_probs = selected_probs + [y_prob]
        trial_mean = np.mean(trial_probs, axis=0)
        trial_score = compute_metrics(y_true_ref, trial_mean, class_names)["macro_f1"]
        if trial_score > best_score + 1e-4:
            selected_rows.append(row)
            selected_probs.append(y_prob)
            best_score = trial_score
            print(f"Added to ensemble: {row['approach']} | val_macro_f1={best_score:.4f}")
        else:
            print(f"Skipped: {row['approach']} | trial_val_macro_f1={trial_score:.4f}")

    return pd.DataFrame(selected_rows), best_score

if CONFIG["run_optional_tta_ensemble"]:
    candidate_df = results_table()
    candidate_df = candidate_df[candidate_df["state_key"].astype(str).str.len() > 0].copy()
    candidate_df = candidate_df[candidate_df["model_family"] != "ensemble"].copy()
    candidate_df = sort_for_validation_selection(candidate_df, top_k=CONFIG["ensemble_top_k"])
    display(candidate_df[["approach", "model_family", "best_val_macro_f1", "params_m", "latency_ms_img", "state_key"]])

    selected_df, ensemble_val_score = greedy_select_ensemble(candidate_df, val_loader)
    if selected_df.empty:
        selected_df = candidate_df.head(1)
        ensemble_val_score = float(selected_df["best_val_macro_f1"].iloc[0])
    print("Selected ensemble members:")
    display(selected_df[["approach", "model_family", "best_val_macro_f1", "state_key"]])

    prob_sum = None
    y_true_ref = None
    n_predictions = 0
    for transform in tta_transforms():
        tta_ds = datasets.ImageFolder(DATA_ROOT / "test", transform=transform)
        tta_ds = filter_excluded_paths(tta_ds, globals().get("EXCLUDED_IMAGE_PATHS", set()), split_name="test TTA")
        tta_loader = DataLoader(
            tta_ds,
            batch_size=CONFIG["batch_size"],
            shuffle=False,
            num_workers=CONFIG["num_workers"],
            pin_memory=device.type == "cuda",
        )
        for _, row in selected_df.iterrows():
            y_true, y_prob = predict_state_key(row["state_key"], tta_loader)
            if y_true_ref is None:
                y_true_ref = y_true
                prob_sum = np.zeros_like(y_prob)
            else:
                assert np.array_equal(y_true_ref, y_true), "TTA loader order mismatch."
            prob_sum += y_prob
            n_predictions += 1

    ensemble_prob = prob_sum / max(n_predictions, 1)
    ensemble_metrics = compute_metrics(y_true_ref, ensemble_prob, class_names)
    print("Validation-greedy ensemble + TTA")
    for key, value in ensemble_metrics.items():
        print(f"{key:>22}: {value:.4f}" if pd.notna(value) else f"{key:>22}: NaN")
    plot_confusion(y_true_ref, ensemble_prob, class_names, "Validation-greedy ensemble + TTA")

    experiment_results.append({
        "approach": f"Validation-greedy ensemble + TTA ({len(selected_df)} models)",
        "model_family": "ensemble",
        "state_key": " + ".join(selected_df["state_key"].tolist()),
        "best_val_macro_f1": ensemble_val_score,
        **ensemble_metrics,
        "params_m": np.nan,
        "trainable_params_m": np.nan,
        "state_dict_mb": np.nan,
        "gz_state_dict_mb": np.nan,
        "sparsity": np.nan,
        "latency_ms_img": np.nan,
        "notes": "Validation-only greedy probability averaging with center-crop TTA on test.",
    })

display(results_table())
```

```python
score_sorted = final_results.sort_values(["macro_f1", "accuracy"], ascending=False).copy()
best_overall_row = score_sorted.iloc[0]

single_model_results = final_results[final_results["model_family"] != "ensemble"].copy()
best_single_row = single_model_results.sort_values(["macro_f1", "accuracy"], ascending=False).iloc[0]

efficiency_candidates = single_model_results.dropna(subset=["params_m", "latency_ms_img"]).copy()
if not efficiency_candidates.empty:
    efficiency_candidates["efficiency_score"] = (
        efficiency_candidates["macro_f1"]
        - 0.005 * np.log1p(efficiency_candidates["params_m"])
        - 0.002 * np.log1p(efficiency_candidates["latency_ms_img"])
    )
    best_efficiency_row = efficiency_candidates.sort_values("efficiency_score", ascending=False).iloc[0]
else:
    best_efficiency_row = best_single_row.copy()
    best_efficiency_row["efficiency_score"] = np.nan

best_by_group = (
    final_results
    .sort_values(["approach_group", "macro_f1", "accuracy"], ascending=[True, False, False])
    .groupby("approach_group", as_index=False)
    .head(1)
    .reset_index(drop=True)
)

baseline_rows = final_results[final_results["approach"].str.startswith(("1a.", "1."))]
if not baseline_rows.empty:
    baseline_macro_f1 = float(baseline_rows.sort_values(["macro_f1", "accuracy"], ascending=False).iloc[0]["macro_f1"])
    final_results["macro_f1_delta_vs_baseline"] = final_results["macro_f1"] - baseline_macro_f1
    best_by_group["macro_f1_delta_vs_baseline"] = best_by_group["macro_f1"] - baseline_macro_f1
else:
    baseline_macro_f1 = np.nan

winner_summary = pd.DataFrame([
    {
        "decision": "Best overall",
        "approach": best_overall_row["approach"],
        "macro_f1": best_overall_row["macro_f1"],
        "balanced_accuracy": best_overall_row["balanced_accuracy"],
        "accuracy": best_overall_row["accuracy"],
    },
    {
        "decision": "Best single model",
        "approach": best_single_row["approach"],
        "macro_f1": best_single_row["macro_f1"],
        "balanced_accuracy": best_single_row["balanced_accuracy"],
        "accuracy": best_single_row["accuracy"],
    },
    {
        "decision": "Best efficiency-aware single model",
        "approach": best_efficiency_row["approach"],
        "macro_f1": best_efficiency_row["macro_f1"],
        "balanced_accuracy": best_efficiency_row["balanced_accuracy"],
        "accuracy": best_efficiency_row["accuracy"],
    },
])

print("Best approach per experiment group:")
group_cols = ["approach_group", "approach", "model_family", "macro_f1", "balanced_accuracy", "accuracy", "best_val_macro_f1"]
if "macro_f1_delta_vs_baseline" in best_by_group.columns:
    group_cols.append("macro_f1_delta_vs_baseline")
display(best_by_group[group_cols].style.format({
    "macro_f1": "{:.4f}",
    "balanced_accuracy": "{:.4f}",
    "accuracy": "{:.4f}",
    "best_val_macro_f1": "{:.4f}",
    "macro_f1_delta_vs_baseline": "{:+.4f}",
}, na_rep="-"))

print("\nDecision summary:")
display(winner_summary.style.format({"macro_f1": "{:.4f}", "balanced_accuracy": "{:.4f}", "accuracy": "{:.4f}"}))

advanced_mask = final_results["approach"].str.startswith(("3.", "4.", "5.", "6."))
advanced_results = final_results[advanced_mask].copy()
under_target = advanced_results[advanced_results["macro_f1"] < CONFIG["target_macro_f1"]]


def class_difficulty_table(y_true, y_prob, class_names):
    y_pred = np.argmax(y_prob, axis=1)
    report = classification_report(
        y_true,
        y_pred,
        target_names=class_names,
        output_dict=True,
        zero_division=0,
    )
    cm = confusion_matrix(y_true, y_pred)
    rows = []
    for idx, cls_name in enumerate(class_names):
        rows.append({
            "class": cls_name,
            "precision": report[cls_name]["precision"],
            "recall": report[cls_name]["recall"],
            "f1": report[cls_name]["f1-score"],
            "support": int(report[cls_name]["support"]),
            "false_negatives": int(cm[idx].sum() - cm[idx, idx]),
            "false_positives": int(cm[:, idx].sum() - cm[idx, idx]),
        })
    return pd.DataFrame(rows).sort_values(["recall", "f1"], ascending=True)

def predictions_for_result_row(row):
    if row["model_family"] == "ensemble":
        if "ensemble_prob" in globals() and "y_true_ref" in globals():
            return y_true_ref, ensemble_prob, row["approach"]
        print("Ensemble predictions are unavailable in memory; falling back to the best single model for class-difficulty analysis.")
        row = best_single_row

    model = build_model_from_family(row["model_family"], NUM_CLASSES).to(device)
    model.load_state_dict(model_state_registry[row["state_key"]]["model_state"])
    y_true, y_prob = collect_predictions(model, test_loader)
    del model
    clear_memory()
    return y_true, y_prob, row["approach"]

y_difficulty, p_difficulty, difficulty_source = predictions_for_result_row(best_overall_row)
difficulty_df = class_difficulty_table(y_difficulty, p_difficulty, class_names)
print(f"\nMost difficult classes based on: {difficulty_source}")
display(difficulty_df.style.format({"precision": "{:.4f}", "recall": "{:.4f}", "f1": "{:.4f}"}))

selected_transfer_label = globals().get("best_transfer_name", "the selected transfer winner")
```

### 15.2.8. Relationships consistently observed across the studies

#### Ensemble learning

All ensemble-oriented studies showed that combining diverse models generally improved performance.

In the project:

* best individual model Macro F1: **0.9585**,
* ensemble Macro F1: **0.9727**,
* improvement: approximately **1.43 percentage points**.

FLANNEL improved Macro F1 from **0.7445** for its strongest individual base learner to **0.7910** for the final ensemble.

The CheXpert study improved mean AUC from **0.894** for its best individual model to **0.940** for the six-model ensemble.

Rajaraman et al. also found that the weighted ensemble of pruned models produced the strongest final result.

#### Data augmentation and TTA

The reviewed studies show that augmentation is not universally beneficial.

* Nishio et al. obtained the best result by combining conventional augmentation, MixUp, and appropriate layer freezing.
* BayTTA showed that standard TTA can sometimes lower performance.
* In the project, simple baseline augmentation reduced Macro F1 from **0.9239 to 0.9141**.
* The optimized augmentation pipeline improved the individual-model Macro F1 to **0.9585**.
* The final ensemble with TTA increased the result further to **0.9727**.

Therefore, augmentation must be mild, anatomically safe, and validated experimentally. Excessive rotations, translations, contrast modifications, or crops may distort medically important structures.

#### Transfer learning

Transfer learning was useful but was not always the strongest strategy.

In the project:

* optimized scratch CNN: Macro F1 = **0.9585**,
* ResNet50 transfer learning: Macro F1 = **0.9447**.

Nishio et al. similarly found that the simpler VGG16 model outperformed ResNet50, DenseNet121, MobileNet, and EfficientNet on a small dataset.

These results suggest that ImageNet pretraining may not always transfer optimally to grayscale radiographs. Chest-X-ray-specific or self-supervised pretraining may provide more relevant features.

#### Pruning

The project showed that conservative pruning can preserve performance:

* source model Macro F1: **0.9512**,
* 4% pruned model Macro F1: **0.9511**.

However, the practical efficiency gain was small. Rajaraman et al. achieved much larger reductions in trainable parameters while maintaining or improving performance. Structured and iterative pruning therefore remains an important area for further work.

#### Difficult pneumonia subtypes

Across the reviewed literature and this project, distinguishing related pneumonia categories was consistently more difficult than recognizing Normal or Tuberculosis cases.

For the project ensemble:

| Class               |         F1 | False negatives | False positives |
| ------------------- | ---------: | --------------: | --------------: |

### 15.2.9. The project vs. The Literature

#### Cases in which the project obtained higher numerical results

The project obtained higher reported metrics than:

* DenResCov-19 on all three- and four-class datasets,
* FLANNEL for every overlapping class and Macro F1,
* the VGG16 augmentation study in accuracy and COVID-19 recall,
* the BayTTA chest X-ray experiments in accuracy and F1,
* most individual transfer-learning and baseline models reported in the reviewed studies.

These comparisons are particularly encouraging because the project used six classes and maintained similar accuracy, balanced accuracy, precision, recall, and Macro F1. This indicates that the result was not caused only by domination of a large or easy class.

#### Cases in which the project did not obtain a higher result

The project did not exceed:

* the **0.9821 F1** obtained by DenResCov-19 in its binary Pneumonia-versus-Healthy task,
* the **0.9901 accuracy and F-score** obtained by the three-class iteratively pruned weighted ensemble.

The likely reasons are:

1. the literature tasks contained fewer classes,
2. the project had to distinguish bacterial from non-COVID viral pneumonia,
3. the literature used modality-specific chest X-ray pretraining,
4. the literature used more aggressive structured pruning,
5. the literature used learned or manually weighted ensemble strategies,
6. the project used a single main dataset rather than additional pretraining cohorts.

#### Cases in which no direct conclusion is valid

The project AUC was numerically higher than the CheXpert study, but the project cannot be claimed to outperform it scientifically because CheXpert involved:

* more than 200,000 images,
* multi-label prediction,
* uncertain labels,
* independent hidden testing,
* radiologist consensus,
* comparison with human specialists.

Thus, the project demonstrated stronger within-dataset numerical metrics, while the CheXpert study demonstrated stronger external and clinical validation.

### Final literature-comparison conclusion

The project produced **highly competitive results** compared with the reviewed literature. Its validation-greedy three-model ensemble achieved **0.9727 Macro F1**, **0.9724 accuracy**, **0.9728 balanced accuracy**, and **0.9974 Macro ROC-AUC** on a six-class test set.

The project numerically exceeded the multiclass results reported by DenResCov-19, FLANNEL, the VGG16 augmentation study, and the BayTTA chest X-ray experiments. It also achieved substantially better F1 scores for COVID-19, bacterial pneumonia, viral pneumonia, and Normal than FLANNEL.

Overall, the project results are especially strong considering the number of classes and the difficulty of separating bacterial from viral pneumonia. However, the results remain an **internal evaluation on one primary dataset**.

## 15.3. Final Conclusions

The main objective of this project was to develop, optimize, and compare convolutional neural network models for the automatic classification of chest X-ray images into six categories: COVID-19, Emphysema, Normal, Pneumonia-Bacterial, Pneumonia-Viral, and Tuberculosis. In addition to classification quality, the study considered practical efficiency indicators, including the number of parameters, checkpoint size, model sparsity, and inference latency. The completed experiments demonstrated that high predictive performance can be achieved without relying exclusively on the largest pretrained architectures.

The ChestX6 dataset initially contained 18,036 images, including 14,551 training images, 1,748 validation images, and 1,737 test images. The dataset images used in the experiments were processed as grayscale 224 × 224 images and converted to three-channel inputs. They were normalized using ImageNet statistics to support both custom and pretrained CNN architectures. An important part of data preparation was the duplicate analysis. A total of 37 duplicate groups were detected, including 17 groups crossing dataset subsets. To reduce the risk of data leakage, 23 lower-priority images were excluded, and the final verification confirmed that no cross-split duplicate leakage remained.

All model-selection decisions were based on validation Macro F1-score. The test set was used only after the architecture, checkpoint, fine-tuning strategy, pruning level, and ensemble members had been selected. This protocol reduced the risk of adapting the final solution to the test data and made the comparison between approaches more reliable. Macro F1 was used as the main selection metric because it assigns equal importance to all six classes and is therefore more informative than accuracy alone when evaluating differences in class difficulty.

The simple CNN trained from scratch established a strong initial reference result, achieving an accuracy of 0.9229 and a Macro F1-score of 0.9239 with only approximately 1.21 million parameters and an inference latency of 1.16 ms per image. Applying standard augmentation and TTA to the same architecture did not improve its overall performance. Its Macro F1 decreased to 0.9141, mainly because bacterial pneumonia recall became lower. This result showed that augmentation is not automatically beneficial and must be adapted to both the medical image characteristics and the selected architecture.

The strongest individual model was the Optimized Baseline CNN trained from scratch with X-ray-safe augmentation, MixUp, regularization, a polishing stage, and TTA. It achieved an accuracy of 0.9585, balanced accuracy of 0.9591, Macro F1-score of 0.9585, and ROC-AUC of 0.9957. The model contained approximately 12.34 million parameters and required about 5.22 ms per image. It substantially improved the recognition of both bacterial and viral pneumonia and outperformed all ImageNet-pretrained models. It was consequently selected as both the best individual model and the best efficiency-aware solution in the project.

Among the transfer-learning architectures, ResNet50 produced the best results. Its staged head warm-up, last-block adaptation, and low-learning-rate polishing resulted in a test Macro F1-score of 0.9447 and an accuracy of 0.9442. DenseNet121 reached a Macro F1-score of 0.9172, EfficientNet-B0 obtained 0.9068, and MobileNetV2 achieved 0.9048. MobileNetV2 was the smallest and fastest transfer-learning architecture, whereas DenseNet121 had the highest inference latency despite its lower predictive performance. These experiments demonstrated that a larger pretrained model does not necessarily provide the best quality–efficiency trade-off and that ImageNet features are not always optimally adapted to grayscale radiographic images.

Additional fine-tuning was performed only for ResNet50 because it was the best transfer model according to validation Macro F1. Conservative refinement did not exceed the previous validation result, so the validation guard retained the stronger original transfer-learning checkpoint. This prevented a weaker fine-tuned state from replacing the better model. The subsequent extra-soft augmentation stage was more successful and improved the ResNet50 Macro F1-score from 0.9447 to 0.9512. It therefore became the strongest transfer-based model, although it still remained below the optimized scratch CNN while using more parameters, a larger checkpoint, and a higher inference latency.

Quality-first pruning was applied to the extra-soft augmented ResNet50. A pruning level of 4% produced a Macro F1-score of 0.9511, which was almost identical to the unpruned result of 0.9512. The model therefore gained measurable sparsity without a meaningful loss of classification quality. However, because global unstructured pruning preserved the original dense tensor shapes, the total number of parameters and raw checkpoint size did not decrease. The compressed checkpoint size decreased only from approximately 92.93 MB to 90.89 MB, while inference latency improved slightly from 6.72 to 6.64 ms per image. The experiment confirmed that mild pruning can preserve quality, but it also showed that unstructured sparsity alone provides limited practical acceleration on standard dense hardware.

The best overall result was achieved by the validation-greedy ensemble with TTA. The final ensemble combined the Optimized Baseline CNN, the 4% pruned ResNet50, and the MobileNetV2 transfer model. These models were selected because their averaged validation predictions improved Macro F1, whereas additional, more similar ResNet50 variants did not provide further benefits. After probability averaging across three models and three TTA transformations, the ensemble achieved an accuracy of 0.9724, balanced accuracy of 0.9728, Macro precision of 0.9730, Macro recall of 0.9728, Macro F1-score of 0.9727, and ROC-AUC of 0.9974. Compared with the basic CNN baseline, this represented an improvement of 0.0488 in Macro F1. Compared with the strongest individual model, the ensemble improved Macro F1 by approximately 1.43 percentage points.

The ensemble produced very high and balanced class-level performance. Tuberculosis was classified perfectly, while Normal, COVID-19, and Emphysema achieved F1-scores close to or above 0.98. The most difficult categories remained Pneumonia-Bacterial and Pneumonia-Viral, with F1-scores of 0.9472 and 0.9496, respectively. Most remaining errors involved bacterial pneumonia being classified as viral pneumonia or the reverse. This confirms that the greatest limitation was not general disease detection but the separation of clinically and radiologically similar pneumonia subtypes.

The final model choice depends on the intended application. The validation-greedy ensemble with TTA is the preferred solution when maximum classification performance is the main objective. However, it requires repeated inference through three separate models and several transformed versions of every test image, making it more computationally expensive. The Optimized Baseline CNN is a more practical solution for deployment because it offers the best individual performance, substantially lower complexity than the ResNet50 models, a smaller checkpoint, and lower latency. The simple CNN may still be appropriate for highly resource-limited environments, but this efficiency is obtained at the cost of lower classification quality.

Further development should include external validation on independent datasets, repeated patient-level cross-validation, confidence intervals, and chest-X-ray-specific or self-supervised pretraining. The distinction between bacterial and viral pneumonia could be improved through hierarchical classification, lung segmentation, attention mechanisms, higher-resolution lung crops, and additional clinically verified images. The ensemble and TTA procedure could be extended with learned or Bayesian probability weighting and explicit uncertainty estimation. Practical deployment could be improved through structured channel pruning, quantization, knowledge distillation, or compression of the ensemble into a smaller student model.

In conclusion, the project successfully developed and evaluated a complete deep-learning pipeline for six-class chest X-ray classification. It showed that carefully designed scratch CNNs can outperform commonly used pretrained architectures, that augmentation must be validated rather than assumed to improve performance, and that conservative pruning can preserve predictive quality. Most importantly, the results demonstrated that combining architecturally diverse models with validation-controlled ensemble selection and TTA produced the highest and most balanced classification performance. The final ensemble achieved a Macro F1-score of 0.9727, while the Optimized Baseline CNN provided the strongest balance between accuracy and computational efficiency.
